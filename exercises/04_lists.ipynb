## 1. What Is a List?

A **list** is an ordered, mutable collection of values enclosed in square brackets `[...]` and separated by commas.

Lists allow us to store, organize, and process multiple related items under a single variable name, eliminating the need to declare repetitive scalar variables.

### Problem: Managing Separate Variables

When tracking multiple biological samples, creating isolated variables becomes unmanageable:

```python
sample_1 = "S001"
sample_2 = "S002"
sample_3 = "S003"

### Solution: Storing Sequences in a List
By grouping related data into a single list, we can pass, inspect, and iterate over the entire dataset efficiently:

```python
sample_ids = ["S001", "S002", "S003"]

print("Contents:", sample_ids)
print("Data type:", type(sample_ids))
print("Number of elements:", len(sample_ids))

### Key Attributes of Python Lists

1. **Ordered:** Elements maintain the exact order in which they are inserted.
2. **Heterogeneous:** A single list can hold mixed data types (`str`, `int`, `float`, `bool`).
3. **Dynamic:** Items can be added, modified, or removed at runtime.

## 6. Slicing Lists

Slicing extracts a contiguous sub-list from an existing list without modifying the original container.

The standard syntax is:
`list[start:stop:step]`

- `start`: The index where the slice begins (inclusive). Defaults to `0` if omitted.
- `stop`: The index where the slice ends (**exclusive**). Defaults to `len(list)` if omitted.
- `step`: The stride or interval between elements. Defaults to `1` if omitted.

### Basic Slicing Mechanics

````python
samples = ["S001", "S002", "S003", "S004", "S005", "S006"]

print("samples[:3]   (First 3 elements): ", samples[:3])
print("samples[2:]   (Index 2 to end):   ", samples[2:])
print("samples[1:4]  (Indices 1, 2, 3):  ", samples[1:4])
print("samples[::-1] (Reversed list):    ", samples[::-1])
````

## 9. Mutability & Variable References

A **mutable** object is one whose internal state can be altered in place after creation without changing its memory identity. 

Lists are mutable; strings and integers are immutable. While mutability allows for efficient in-place updates, it introduces a subtle behavior when binding multiple variable names to the same list container.


### In-Place Element Mutation

Direct item assignment modifies the underlying list object directly:

````python
samples = ["S001", "S002", "S003"]
print("Initial samples:", samples)

samples[0] = "S100"
print("Mutated samples:", samples)
````

### The Shared Reference Experiment (Aliasing)

Observe what happens when we assign an existing list variable to a new variable name:

````python
original = ["A", "B", "C"]

# Create an alias pointing to 'original'
alias = original

# Mutate the list through 'alias'
alias[0] = "X"

print("original:", original)
print("alias:   ", alias)
````

### Conceptual Investigation

> **Think About It:**
> 1. Why did modifying `alias[0]` also change the contents printed by `original`?
> 2. Did `alias = original` create a brand-new copy of the list, or did it assign a second label to the exact same list object?
> 3. How can you verify using Python's built-in `id()` function whether two variables point to the same memory identity?

````python
# Verification using memory address inspection:
print("Memory ID of original:", id(original))
print("Memory ID of alias:   ", id(alias))
print("Are they identical?   ", id(original) == id(alias))
````

### Key Takeaway: Variable Names are Labels, Not Boxes

In Python, variables do not "contain" data directly; they act as **pointers or reference labels** attached to objects in memory.

$$\text{original} \longrightarrow \begin{bmatrix} \text{"X"}, & \text{"B"}, & \text{"C"} \end{bmatrix} \longleftarrow \text{alias}$$

- **Aliasing:** Writing `alias = original` creates two references pointing to a **single** list in memory. Any mutation through one label is visible through the other.
- **Copying:** To create an independent duplicate list that can be mutated safely without affecting the original, use explicit slicing (`copy_list = original[:]`) or `.copy()` (`copy_list = original.copy()`).

#### My Analysis

1. **Explanation of alias mutation behavior:** Writing `alias = original` does not create a new list or allocate new memory; it simply binds a second variable label (`alias`) to the exact same list object in memory. Because both variable names point to the same memory identity (`id(original) == id(alias)`), modifying the list through `alias[0] = "X"` mutates the shared underlying object directly, making the change visible when accessing `original`.
2. **How to safely duplicate a list before modifying it:** To create an independent duplicate (a shallow copy) that can be safely mutated without altering the original list, use one of the following explicit copying techniques:

    - List slicing: `copy_list = original[:]`

    - The `.copy()` method: `copy_list = original.copy()`

    - The `list()` constructor: c`opy_list = list(original)`

````python
original = ["A", "B", "C"]

# Create an independent shallow copy
independent_copy = original.copy()

# Mutate the copy
independent_copy[0] = "X"

print("original:        ", original)  # Remains ['A', 'B', 'C']
print("independent_copy:", independent_copy)  # Becomes ['X', 'B', 'C']
print("Do they share memory ID?", id(original) == id(independent_copy))  # False
````

## 13. Sorting Lists

Sorting arranges sequence values into a defined order. Python provides two distinct ways to sort:

* `sorted(list)`: Function that returns a new sorted list, leaving the original untouched.
* `list.sort()`: Method that mutates the existing list in-place and returns `None`.

````python
values = [14, 2, 19, 7, 11]

# Out-of-place sorting
sorted_values = sorted(values)
print("Original values after sorted():", values)
print("Returned sorted list:         ", sorted_values)
print("-" * 45)

# In-place sorting
values.sort()
print("Original values after .sort(): ", values)
````

### Investigation: sorted() vs .sort()

#### Questions for Reflection
1. Does `sorted()` modify the original list?
2. Does `.sort()` modify the original list?
3. What is the return value of `.sort()` vs `sorted()`?

#### My Analysis

1. **Does `sorted()` modify the original list?**  
   **No.** `sorted()` creates and returns a brand-new list object in memory, leaving the original list (`[14, 2, 19, 7, 11]`) untouched.

2. **Does `.sort()` modify the original list?**  
   **Yes.** `.sort()` mutates the existing list directly in memory (*in-place*).

3. **What is the return value of `.sort()` vs `sorted()`?**  
   * `sorted(values)` returns the **new sorted `list`** (`[2, 7, 11, 14, 19]`).
   * `values.sort()` returns **`None`**.

> **Critical Trap:** Never write `values = values.sort()`. Because `.sort()` returns `None`, assigning its output back to the original variable name overwrites and destroys your dataset!

## 14. Sorting and Data Types

Data type directly governs sort behavior. Numbers sort numerically, while text strings sort lexicographically (character by character according to ASCII values).

````python
# Numeric sort
numbers = [2, 10, 3, 25]
numbers.sort()
print("Sorted integers:", numbers)

# String sort
string_numbers = ["2", "10", "3", "25"]
string_numbers.sort()
print("Sorted strings: ", string_numbers)
````

### Investigation: The Lexicographical Trap

#### Question for Reflection
Why does `"10"` sort before `"3"` when stored as strings?

#### My Analysis

1. **Explanation of string comparison mechanics:**  
   Strings are sorted **lexicographically (character-by-character from left to right)** based on their ASCII/Unicode values, rather than as complete numerical quantities. When Python compares `"10"` and `"3"`, it evaluates the first character of each string: `'1'` vs. `'3'`. Because `'1'` has a lower ASCII code ($49$) than `'3'` ($51$), `"10"` is placed before `"3"`, completely ignoring the numeric value of the full string.

2. **Why casting strings to `int` or `float` before sorting is critical:**  
   Explicitly converting string inputs into numeric types (`int()` or `float()`) shifts Python from character-based ASCII evaluation to true **mathematical evaluation**. Without type casting, multi-digit numeric strings will sort out of order (e.g., `"100"` comes before `"2"`), leading to corrupted rankings, invalid threshold sorting, and broken experimental reports.

## 17. Custom Sorting with `key=` and `lambda`

In Section 16, we saw that Python sorts tuples based on the element at index `0`. Swapping the order in `zip()` allowed us to change what Python sorts by.

However, rebuilding or re-zipping lists every time we want to sort by a different attribute is inefficient and clunky. 

Instead, Python allows us to keep our data structured as `(student, grade)` or `(sample_id, concentration)` and pass a **custom sorting key** using the `key=` parameter and a `lambda` expression.

---

### 17.1 What Is a `lambda` Expression?

A `lambda` expression is a small, anonymous (unnamed) function defined on a single line:

```text
    Key argument in sort
         │
  key = lambda record: record[1]
                 │        │
  Input parameter        Extracted value used as the sorting criteria
  (each item in list)    (e.g., index 1)

### 17.2 Sorting (`name`, `grade`) Records Without Re-zipping

````python
students = ["Ava", "Ben", "Carl", "Dana"]
grades = [85, 92, 78, 100]

# Standard structure: (Name at index 0, Grade at index 1)
pairs = list(zip(students, grades))
print("Original paired data:           ", pairs)

# Default Sort (Sorts by index 0 — Name)
by_name = sorted(pairs)
print("Sorted by Name (default index 0):", by_name)

# Custom Sort with key=lambda (Sorts by index 1 — Grade)
by_grade = sorted(pairs, key=lambda record: record[1])
print("Sorted by Grade (custom index 1):", by_grade)
````

### 17.3 Sorting in Descending Order

````python
top_grades = sorted(pairs, key=lambda record: record[1], reverse=True)
print("Ranked by Grade (Highest First):", top_grades)
````

````python
# Raw multi-line assay output stream
assay_raw = """S001,S002,S003,S004
15.2,8.7,100.4,12.0"""

# Step 1: Split string into rows and row elements
rows = assay_raw.strip().split("\n")
sample_ids = rows[0].split(",")
raw_concentrations = rows[1].split(",")

# Step 2: Convert concentration strings to floats using a loop
concentrations = []
for val in raw_concentrations:
    concentrations.append(float(val))

# Step 3: Zip IDs and concentrations into paired tuples: (ID, Concentration)
samples = list(zip(sample_ids, concentrations))

# Step 4: User interaction
print("Select Sort Criteria:")
print("1: Sort by Concentration\n2: Sort by Sample ID")
choice = input("Enter choice (1 or 2): ")

# Step 5: Perform custom in-place sorting using key=lambda
if choice == "1":
    # Sort in-place by concentration (index 1)
    samples.sort(key=lambda record: record[1])
elif choice == "2":
    # Sort in-place by sample ID (index 0)
    samples.sort(key=lambda record: record[0])

# Step 6: Display formatted results
print("\n--- Sorted Assay Report ---")
for sample_id, conc in samples:
    print(f"Sample: {sample_id:<6} | Concentration: {conc:>6.2f} µM")
````

Build an automated, terminal-driven pipeline that ingests raw multi-line assay exports, parses multi-attribute biological records into parallel data containers, computes statistical metrics manually, and executes interactive custom sorting via `key=` and `lambda`.

---

### Project Scope & Deliverables

Your program must process raw tabular data containing:
- **Sample IDs** (e.g., `"S001"`)
- **Organisms** (e.g., `"E_coli"`)
- **Concentrations ($\mu\text{M}$)** (e.g., `12.5`)

#### System Requirements:
1. **Multi-Stage Text Parsing:** Split raw multi-line CSV text into rows, then split each row into constituent attribute fields.
2. **Type Casting:** Convert string representations of numerical concentrations into `float` values using loops.
3. **Coupling Collections:** Combine parallel lists into paired record tuples (`(sample_id, organism, concentration)`) using `zip()`.
4. **Interactive Sorting Engine:** Prompt the user (`input()`) to select sorting criteria:
   - Option `1`: Sort by **Concentration** (numerical order).
   - Option `2`: Sort by **Sample ID** (lexicographical order).
   - Apply `.sort(key=lambda ...)` to execute the selection in-place.
5. **Statistical Metrics:** Manually compute the **mean concentration** and locate the **sample record with the highest concentration** without using NumPy, Pandas, `sum()`, or `max()`.
6. **Formatted Reporting:** Display a clean tabular summary report.

### Technical Constraints

````python
# ==============================================================================
# SECTION 19: INTEGRATED CHALLENGE — LABORATORY SAMPLE MANAGER (SOLUTION)
# ==============================================================================

raw_assay_data = """S001,E_coli,12.5
S002,E_coli,8.7
S003,S_aureus,22.1
S004,S_aureus,15.4
S005,P_aeruginosa,31.0
S006,P_aeruginosa,19.8
S007,E_coli,5.2
S008,S_aureus,11.3"""

# ------------------------------------------------------------------------------
# STEP 1 & 2: Multi-Stage Parsing & Explicit Type Conversion
# ------------------------------------------------------------------------------
rows = raw_assay_data.strip().split("\n")

sample_ids = []
organisms = []
concentrations = []

for row in rows:
    # Unpack comma-separated string fields directly
    sid, org, conc_str = row.split(",")

    # Store attributes in persistent parallel collections
    sample_ids.append(sid)
    organisms.append(org)
    concentrations.append(float(conc_str))  # Type cast to float

# ------------------------------------------------------------------------------
# STEP 3: Coupling Collections into Record Tuples
# ------------------------------------------------------------------------------
# Form paired tuples: (sample_id, organism, concentration)
records = list(zip(sample_ids, organisms, concentrations))
print(records)

# ------------------------------------------------------------------------------
# STEP 4: Interactive Sorting Engine
# ------------------------------------------------------------------------------
print("Select Sorting Option:")
print("1: Concentration (Numerical)")
print("2: Sample ID (Lexicographical)")
choice = input("Enter choice (1 or 2): ")

if choice == "1":
    # Sort in-place by concentration (index 2) descending (highest yield first)
    records.sort(key=lambda record: record[2], reverse=True)
elif choice == "2":
    # Sort in-place by sample ID (index 0) ascending
    records.sort(key=lambda record: record[0])

# ------------------------------------------------------------------------------
# STEP 5: Manual Statistical Accumulation
# ------------------------------------------------------------------------------
total_sum = 0.0
max_record = records[0]  # Full tuple initialized to track ID + Org + Conc

for record in records:
    conc = record[2]

    # Accumulate total sum manually
    total_sum += conc

    # Track maximum sample record manually
    if conc > max_record[2]:
        max_record = record

# Compute mean concentration
mean_conc = total_sum / len(records)

# ------------------------------------------------------------------------------
# STEP 6: Formatted Summary Report
# ------------------------------------------------------------------------------
print("\n" + "=" * 55)
print(f"{'SAMPLE MANAGER REPORT':^55}")
print("=" * 55)
print(f"{'Sample ID':<10} | {'Organism':<15} | {'Concentration':>12}")
print("-" * 55)

for sid, org, conc in records:
    print(f"{sid:<10} | {org:<15} | {conc:>9.2f} µM")

print("=" * 55)
print(f"Total Samples Processed: {len(records):>24d}")
print(f"Mean Concentration:      {mean_conc:>21.2f} µM")
print(
    f"Peak Yield Record:       {max_record[0]} ({max_record[1]}) @ {max_record[2]:.2f} µM"
)
print("=" * 55)
````

### 20.1 Core Concepts & Mechanics

#### Lists & Strings
1. **What is a list?**
   * **Answer:** A list is a ordered, mutable sequence container in Python capable of storing dynamic collections of heterogeneous items (integers, floats, strings, or nested lists) under a single variable reference.

2. **How is a list different from a string?** (Consider mutability, element data types, and structural flexibility.)
   * **Answer:** Strings are immutable sequences of single text characters where elements cannot be modified in-place after creation. Lists are mutable containers that allow element insertion, deletion, and direct index modification without re-allocating the entire container.

3. **Why are lists indispensable for biological data processing?**
   * **Answer:** Biological datasets (genomic sequences, assay values, sample metadata) are dynamic, sequential, and multi-attribute. Lists enable manual parsing, memory-efficient updates, element tracking, and customized sorting without requiring external data libraries.

#### Indexing & Slicing
1. **How does zero-based list indexing work? How does negative indexing access elements relative to the end of a list?**
   * **Answer:** Zero-based indexing assigns the first element index `0`, incrementing up to `len(list) - 1`. Negative indexing counts backward from the right end, where `-1` points to the last element, `-2` to the second to last, and so forth.

2. **How does list slicing (`[start:stop:step]`) work? What elements are included or excluded by the `stop` boundary?**
   * **Answer:** Slicing extracts sub-sequences using `[start:stop:step]`. The `start` boundary is inclusive, while the `stop` boundary is exclusive (it cuts off immediately before that index). `step` defines iteration direction and interval.
3. **What happens when you attempt to access an index that does not exist?**
   * **Answer:** Attempting to access an unassigned index (e.g., `sample_list[10]` when `len` is 4) raises an `IndexError: list index out of range`.
#### Mutability & Memory References
1. **What does it mean for a list to be *mutable*?** Contrast this with string immutability.
   * **Answer:** Mutability means a data structure's contents can be altered in-place at its existing memory address. Modifying a string creates a brand-new object in memory, while modifying a list mutates the existing memory block.
2. **What happens when two variables refer to the exact same list in memory (aliasing)?** How does mutating the list through one variable affect the second variable?
   * **Answer:** When two variables point to the same list (aliasing via `b = a`), both variables store identical memory address pointers. Mutating the list through variable `b` immediately reflects in variable `a`.

#### List Operations & Methods
1. **What does `.append()` do, and what does it return?**
   * **Answer:** `.append(item)` appends a single object to the end of a list in-place and returns `None`.
2. **What does `.remove()` do when an element is found? What occurs if the value is not present in the list?**
   * **Answer:** `.remove(val)` searches left-to-right and deletes the first matching instance of `val`, shifting remaining elements left. If `val` is missing, it raises a `ValueError`.
3. **How does `.sort()` differ from `sorted()`?** (Focus on in-place mutation vs. returning a new list copy.)
   * **Answer:** `.sort()` mutates the original list in-place and returns `None`. `sorted(iterable)` leaves the original object untouched and allocates a brand-new sorted list copy.

### 20.2 Data Parsing & Structure

#### Structured Data & Text Extraction
1. **How does `.split()` transform a raw delimiter-separated string into a list of sub-strings?**
   * **Answer:** `.split(delimiter)` scans a string, cuts it at each occurrence of `delimiter`, drops the delimiter characters, and returns a list containing extracted substring fields.
2. **How can multi-line string exports be parsed into individual rows using `.split("\n")` or `.splitlines()`?**
   * **Answer:** Multi-line strings are parsed into discrete record strings by splitting on newline characters (`.split("\n")` or `.splitlines()`).

3. **How do you break down individual row strings into distinct column attributes inside a loop?**
   * **Answer:** Inside a `for` loop iterating over row strings, calling `row.split(",")` breaks each row down into individual column field strings, which can then be unpacked into target variables (`id, org, conc = row.split(",")`).

#### Data Types & Lexicographical Traps
1. **Why is explicit type conversion (e.g., `float()`, `int()`) mandatory after parsing numerical data from text strings?**
   * **Answer:** All values extracted via `.split()` are text strings (`str`). Numerical operations and mathematical sorting fail or sort incorrectly unless strings are explicitly cast using `float()` or `int()`.

2. **How does sorting text strings representing numbers (e.g., `["2", "10", "3"]`) differ from sorting actual numerical types?** What is lexicographical order?
   * **Answer:** Text strings sort lexicographically (character-by-character according to ASCII/Unicode values). Under string sorting, `"10"` precedes `"3"` because `'1'` has a lower ASCII value than `'3'`. Numeric sorting evaluates the full scalar quantity `($3 < 10$)`.

### 20.3 Sequence Pairing & Custom Sorting

#### Zipped Collections (`zip()`)
1. **What specific problem does `zip()` solve when handling parallel lists of data?**
   * **Answer:** `zip()` aggregates elements from two or more parallel sequences position-by-position, forming a sequence of tuple pairs (`(item_A, item_B)`).
2. **What happens to remaining elements when sequences passed to `zip()` have unequal lengths?**
   * **Answer:** `zip()` stops constructing tuples as soon as the shortest input sequence is exhausted, quietly truncating excess elements from longer sequences.

3. **How does changing the tuple positional order in `zip(A, B)` versus `zip(B, A)` alter the default sorting behavior of Python?**
   * **Answer:** Python sorts tuples by index `0` first. `zip(grades, students)` sorts primarily by grade, whereas `zip(students, grades)` sorts primarily by student name.

#### Custom Sorting Keys (`key=` & `lambda`)
1. **What role does the `key=` parameter play in `.sort()` and `sorted()`?**
   * **Answer:** The `key=` parameter accepts a function that extracts a specific comparison value from each element, governing how Python orders items without altering the list's structure.
2. **What is a `lambda` expression, and how does `lambda record: record[1]` select specific tuple elements as the primary sort criteria?**
   * **Answer:** A `lambda` expression is a single-line anonymous function. `lambda record: record[1]` takes a list item (`record`) as input and returns the value at index `1` as the evaluation criteria.

3. **How do you combine `key=lambda ...` with `reverse=True` to rank records in descending numerical order?**
   * **Answer:** Combining `key=lambda record: record[2]` with `reverse=True` evaluates the value at index `2` for every item and orders the records from highest to lowest.
