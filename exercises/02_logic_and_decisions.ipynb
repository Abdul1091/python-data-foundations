### Observations & Key Takeaways

### Truthiness & Falsy Values
- **Falsy Values:** `0`, `0.0`, `""`, and `None` all evaluate to `False`. In Python, any numeric zero, empty sequence or string, or `None` represents a blank or uninitialized state and is treated as **falsy**.
- **Truthy Values:** Non-zero numbers (`1`, `-1`), non-empty strings (`"help"`), and initialized objects evaluate to `True`.

### `callable()` Function Analysis
`callable(analyze_sample)` returns `True` while `callable(25)` returns `False` because `analyze_sample` is a function created in memory, whereas `25` is an integer primitive. In Python, functions are first-class objects that implement the internal `__call__()` method, enabling them to be executed using parenthesis syntax `()`, which integer objects do not support.

## 6. `if` Statements

An `if` statement allows a program to execute a block of code conditionally—only when an expression evaluates to `True`.

In Python, the block of code inside an `if` statement is defined by **indentation** (typically 4 spaces).

```python
temperature = 38.2

if temperature > 37.5:
    print("Temperature is elevated.")
```

### Experiment: Changing the State

Now observe what happens when `temperature` is changed to a value below the threshold:

```python
temperature = 36.8

if temperature > 37.5:
    print("Temperature is elevated.")
```

### Important Observation & Reflection

Notice that executing the cell above produces **no output at all**.

**Why?**
1. Python evaluates the condition `temperature > 37.5` (`36.8 > 37.5`), which evaluates to `False`.
2. When the condition in an `if` statement is `False`, Python completely skips the indented block beneath it and continues execution immediately after the block.
3. Because there are no further instructions outside the indented block, the program finishes without printing anything.

## 7. `if / else`

`if` allows a program to handle one condition.

`else` provides an alternative when that condition is `False`.

This creates a two-way branch, guaranteeing that **exactly one** of the two blocks will execute.

```python
temperature = 36.8

if temperature > 37.5:
    print("Temperature is elevated.")
else:
    print("Temperature is not elevated.")
```

## 8. `if / elif / else`

`if` handles the initial condition, `elif` (short for *else if*) allows testing additional mutually exclusive conditions, and `else` catches any remaining cases.

This structure is essential when a measurement falls into more than two categories, such as:
- **Low**
- **Normal**
- **High**

Python evaluates these conditions sequentially from top to bottom. As soon as one condition evaluates to `True`, its block executes, and Python skips all remaining branches.

## 9. Multiple `if` vs `if / elif`

Multiple independent `if` statements and an `if / elif / else` chain do not behave the same way.

The key question to ask when designing control flow is:

> *Can more than one condition be satisfied simultaneously, or are the outcomes mutually exclusive?*

### Experiment 1: Independent `if` Statements

```python
score = 85

if score >= 50:
    print("Pass")

if score >= 70:
    print("Good")

if score >= 80:
    print("Excellent")
```

### Experiment 2: Mutually Exclusive `if / elif` Chain

```python
score = 85

if score >= 50:
    print("Pass")
elif score >= 70:
    print("Good")
elif score >= 80:
    print("Excellent")
```

## 10. Strings in Decisions

Decision logic in Python is not limited to numerical comparisons. Data pipelines and automated workflows frequently make decisions based on string matching, categorical metadata, and text inputs.

### String Equality Matching

The standard comparison operators (`==`, `!=`) work on strings just as they do on numbers.

```python
organism = "Plasmodium falciparum"

if organism == "Plasmodium falciparum":
    print("Malaria parasite identified.")
```

### Interactive Inputs and Case Sensitivity

When receiving input from users or external log files, variations in casing can cause exact equality checks (`==`) to fail unexpectedly.

```python
organism = input("Enter organism name: ")

if organism == "Plasmodium falciparum":
    print("Malaria parasite identified.")
else:
    print("Different organism or case mismatch.")
```

### Refining Logic with String Methods

To make string-based decisions robust against casing variations, normalize strings using methods like `.lower()` or `.strip()` prior to evaluation (connecting directly to string manipulation concepts from Notebook 01).

```python
organism = input("Enter organism name: ")

# Normalize input to lowercase for case-insensitive matching
if organism.lower().strip() == "plasmodium falciparum":
    print("Malaria parasite identified (case-normalized match).")
else:
    print(f"Different organism identified: '{organism}'")
```

## 11. Nested Decisions

A nested decision is a conditional statement placed inside another conditional statement. 

Nested conditions are useful when a secondary check only makes sense if a primary condition evaluates to `True`.

```python
positive = True
concentration = 15.2

if positive:
    if concentration > 10:
        print("Positive sample with high concentration.")
```

### Changing the State

Observe what happens when the primary condition changes to `False`:

```python
positive = False
concentration = 15.2

if positive:
    if concentration > 10:
        print("Positive sample with high concentration.")
```

### Question & Refactoring

> **Question:** Could this same logic be expressed using a single condition with `and`?

Try rewriting the example above without using nesting.

```python
positive = True
concentration = 15.2

if positive and concentration > 10:
    print("Positive sample with high concentration.")
```

### Core Takeaway

**Nested code isn't automatically better code.**

While nesting is necessary when intermediate steps or secondary `else` branches exist, combining conditions with logical operators (`and`/`or`) keeps your code flatter, cleaner, and easier to read—a key principle of writing maintainable Python code.

### Exercise 4: Multi-Parameter Scientific Gatekeeper

A sample is considered suitable for downstream sequencing only when **all three** of the following conditions are simultaneously satisfied:

Write a clean, single-expression control flow check representing this decision rule.

```python
# Exercise 4: Multi-Parameter Scientific Gatekeeper

# Defined thresholds
CONC = 15.0      # ng/µL
MIN_TEMP = 36.0  # °C
MAX_TEMP = 38.0  # °C

# Sample values
conc = 20.0     # ng/µL
temp = 37.9     # °C
qc_result = True

# Individual evaluations
is_conc_valid = conc >= CONC
is_temp_valid = MIN_TEMP <= temp <= MAX_TEMP  # Fixed: temp in the middle
is_qc_valid = qc_result

# Single-expression control flow check
sing_exprssn = is_conc_valid and is_temp_valid and is_qc_valid

print(f"Temperature Valid ({temp}°C): {is_temp_valid}")
print(f"Concentration Valid ({conc} ng/µL): {is_conc_valid}")
print(f"QC Flag Valid: {is_qc_valid}")

print(f"---> Overall check status: {sing_exprssn}")

# You can also get the same result using if/else
if qc_result and (conc >= CONC) and (MIN_TEMP <= temp <= MAX_TEMP):
    print("PASS: Sample is suitable for downstream sequencing.")
else:
    print("FAIL: Sample does not meet quality criteria.")
```
