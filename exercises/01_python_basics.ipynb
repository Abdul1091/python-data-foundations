## 7. String Indexing

Because strings are sequences of characters, individual characters can be
accessed using an index.

Python uses **zero-based indexing**.

```python
sample_id = "BIO2026-SAMPLE-001"

print(sample_id[0])
print(sample_id[1])
print(sample_id[2])

print(sample_id[-1])
print(sample_id[-2])
```

### Observation

**Positive indices:** `0` (first character) to `len(string) - 1` (last character)
**Negative indices:** `-1` (last character) down to `-len(string)` (first character)

## 8. String Slicing

Slicing allows us to extract a portion of a string.

The basic syntax is:

`string[start:stop]`

The `stop` position is excluded.

```python
sample_id = "BIO2026-SAMPLE-001"

print(sample_id[0:3])
print(sample_id[3:7])
print(sample_id[8:14])

# Select every second character
print(sample_id[::2])

# Reverse the entire string
print(sample_id[::-1])
```

### Practice

Using the sample ID above:

- Extract `"BIO"`
- Extract the year
- Extract `"SAMPLE"`
- Extract the numerical identifier
- Reverse the entire string

For each operation, explain the slice you used.

```python
sample_id = "BIO2026-SAMPLE-001"

# 1. Extract 'BIO'
print(sample_id[0:3])

# 2. Extract the year
print(sample_id[3:7])

# 3. Extract 'SAMPLE'
sample_id[8:14]

# 4. Extract the numerical identifier
# print(len(sample_id))
# print(sample_id[15:])
print(sample_id[-3:])

# 5. Reverse the entire string
print(sample_id[::-1])
```

### Explanation of Slices Used

* **`sample_id[0:3]`**: Starts at index 0 and stops before index 3 (extracts indices 0, 1, 2 → `"BIO"`).
* **`sample_id[3:7]`**: Starts at index 3 and stops before index 7 (extracts indices 3, 4, 5, 6 → `"2026"`).
* **`sample_id[8:14]`**: Starts at index 8 (after the first hyphen) and stops before index 14 (extracts `"SAMPLE"`).
* **`sample_id[-3:]`**: Uses negative indexing starting 3 positions from the end (index -3) and leaves `stop` empty to slice all the way to the end of the string (extracts `"001"`).
* **`sample_id[::-1]`**: Leaving `start` and `stop` empty defaults to the whole string, while `step = -1` steps backward one character at a time, reversing the string entirely (extracts `"100-ELPMAS-6202OIB"`).

## 9. String Methods

Python provides methods for transforming and working with strings.

Some useful methods include:

- `.upper()`
- `.lower()`
- `.replace()`
- `.strip()`
- `.split()`

```python
organism = "Plasmodium falciparum"

print(organism.upper())
print(organism.lower())
```

```python
sample_id = "BIO2026-SAMPLE-001"

print(sample_id.replace("SAMPLE", "CONTROL"))
```

```python
sample_name = "   malaria sample   "

print(sample_name)
print(sample_name.strip())
```

```python
organism = "Plasmodium falciparum"

parts = organism.split()

print(parts)
# .split() returns a list
print(parts[0])
print(parts[1])
```

```python
# Methods don't necessarily modify the original string
organism = "Plasmodium falciparum"

organism.upper() # Python creates "PLASMODIUM FALCIPARUM" in memory, but it gets immediately discarded!

print(organism)

organism = organism.upper() # Captures the new uppercase string and rebinds 'organism' to it

print(organism)
```

## 10. String Immutability

Strings in Python are immutable.

This means that once a string object has been created, its individual
characters cannot be changed directly.

```python
sample_id = "BIO2026"
sample_id[0] = "X"
```

### Observation

Python does not allow individual characters in a string to be reassigned.

Instead, I can create a new string.

```python
sample_id = "BIO2026"

new_sample_id = "X" + sample_id[1:]

print(new_sample_id)
```

### Key Concepts Check

#### Variables
- **What is a variable?** A named reference (or label) that points to a value stored in computer memory.
- **What happens when a variable is reassigned?** The variable reference is updated to point to a new value in memory; the old value is eventually cleared by Python's garbage collector.

#### Data Types
- **What is the difference between `int`, `float`, `str`, and `bool`?** 
  - `int`: Whole numbers (e.g., `5`)
  - `float`: Numbers with decimal points (e.g., `12.5`)
  - `str`: Sequence of text characters inside quotes (e.g., `"Plasmodium"`)
  - `bool`: Truth values (`True` or `False`)
- **How can I determine the type of a value?** Use the built-in `type()` function (e.g., `type(sample_id)`).

#### Input & Type Conversion
- **What type does `input()` return?** It *always* returns a string (`str`), regardless of what the user types.
- **Why is type conversion sometimes necessary?** Numerical operations (like addition or multiplication) cannot be performed on raw strings; string inputs must be converted using `int()` or `float()` first.

#### Strings
- **What does zero-based indexing mean?** The first character in a string is located at index `0`, the second at index `1`, and so on.
- **What is the difference between indexing and slicing?** Indexing retrieves a *single character* (e.g., `s[0]`), whereas slicing extracts a *substring range* (e.g., `s[0:3]`).
- **What does `[::-1]` do?** It steps through the string backward from start to end, effectively reversing the string.
- **What does string immutability mean?** Strings cannot be altered after creation. You cannot reassign individual characters in-place (e.g., `s[0] = 'A'` throws a `TypeError`).
- **What is the difference between a string method returning a new value vs. modifying the original?** String methods (like `.upper()`) do not alter the existing string in memory; they construct and return a *new string*, which must be reassigned to a variable to save.
