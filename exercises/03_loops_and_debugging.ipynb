### The Concept of Iteration

**Iteration** means repeating a process or set of instructions systematically. 

Instead of duplicating lines of code, an iteration structure tells the program:
1. **The Core Action:** *What operation to perform on a single item.*
2. **The Sequence or Condition:** *How many times or under what criteria to repeat that action.*
3. **The State Change:** *How the system advances toward completion on every pass.*

By delegating repetition to Python, we move from writing $N$ lines of code for $N$ samples to writing a single, reusable block of logic that scales effortlessly from 5 samples to 10,000,000.

## 3. Loop Conditions & Boundary Experiments

A `while` loop's behavior is dictated entirely by the interaction between its **initial state**, **continuation condition**, and **update step**.

### Base Loop Example

Observe how augmented assignment (`+=`) simplifies state updates (`sample_number += 1` is shorthand for `sample_number = sample_number + 1`).

````python
sample_number = 1

while sample_number <= 3:
    print(f"Processing sample {sample_number}")
    sample_number += 1
````

### The Boundary Zero-Execution Case

What happens if the starting value fails the condition immediately before entering the loop?

````python
sample_number = 10

while sample_number <= 3:
    print(f"Processing sample {sample_number}")
    sample_number += 1

print("Loop finished (or was skipped entirely).")
````

### Key Architectural Insight

> **Zero-Execution Behavior:** If the continuation condition evaluates to `False` on the very first check, the loop body is **completely skipped**. 

This is a feature, not a bug: it guarantees that a program won't attempt to process items when pre-conditions aren't met (for example, attempting to process 0 pending samples in a queue).

#### My Reflection

1. **What happens when `start = 10` and `condition = sample_number <= 3`?** `The loop body is skipped entirely because 10 <= 3 evaluates to False on the very first check. The program immediately jumps to the code below the loop, printing only "Loop finished (or was skipped entirely)." and leaving sample_number at 10.`
2. **Why is zero-execution useful in automated pipelines?** `It serves as an automatic guard clause that prevents empty, unnecessary, or unsafe processing when inputs fail initial requirements—such as receiving a sample queue of size 0, an uninitialized batch, or an out-of-bounds temperature before data analysis begins.`

## 4. Avoiding Infinite Loops

A `while` loop can continue forever if its condition never becomes `False`.

For a loop to terminate, something involved in the condition must eventually change in a way that makes the condition evaluate to `False`.

### Case Study: Missing State Mutation

````python
# DO NOT RUN - INFINITE LOOP EXAMPLE
# count = 1
# while count <= 5:
#     print(count)
````

### Why is this an infinite loop?

1. `count` starts at `1`
2. `count <= 5` evaluates to `True`
3. The loop body executes
4. `count` is never updated or mutated inside the body
5. Therefore, `count` remains `1` forever, and the condition stays `True` indefinitely

The loop has **no path toward termination**.

### Fixing the Path to Termination

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### Debugging Challenge: Directional Logic Error

Consider the following loop logic:

```python
count = 10

while count > 0:
    print(count)
    count += 1
Critical Analysis Question:

Why will this loop never terminate?

What specific code change allows it to terminate properly?

#### My Analysis

1. **Why it never terminates:** `count starts at 10, and the condition checks if count > 0. Inside the loop, count += 1 increments count upwards (10, 11, 12, ...), moving it further away from $0$ with every iteration. Because count will always remain greater than $0$, the condition count > 0 stays True indefinitely.`
2. **Required fix:** `Change the state update from addition to subtraction by decrementing count (count -= 1). This moves count downwards towards $0$ (10, 9, 8, ...), eventually making count > 0 evaluate to False when count hits $0$.`

## 5. `for` Loops

A `for` loop is used when we want to iterate through a sequence or another iterable object.

Instead of manually maintaining state, conditions, and explicit update steps, Python automatically yields the next item in the sequence on each pass until the sequence is exhausted.

### Basic Sequence Traversal

````python
for sample in range(1, 6):
    print(f"Processing sample {sample}")
````

### Comparison: `while` vs. `for`

#### `while` Loop Approach

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

````python
for count in range(1, 6):
    print(count)
````

### Architectural Comparison

In the `while` version, we must explicitly control:
* **Initialization:** `count = 1`
* **Continuation Condition:** `count <= 5`
* **State Mutation:** `count += 1`

In the `for` version, Python handles the iteration automatically using the values supplied by `range(1, 6)`.

#### My Comparison & Reflection

1. **When is a `for` loop naturally suited?** `When you know the number of iterations in advance or when iterating over a finite collection/sequence of known size (such as a list of sample IDs, a range of numbers, string characters, or dictionary keys).`
2. **When is a `while` loop naturally suited?** `When the number of iterations is unknown in advance and processing must continue until a specific condition or event occurs (such as waiting for a sensor temperature to stabilize, prompting a user until valid input is received, or monitoring a reaction until a threshold is hit).`

## 6. `range()`

The `range()` function generates an immutable sequence of numbers on demand, making it the primary driver for count-controlled `for` loops.

### Syntax Variations

The `range()` function accepts up to three arguments:

- `range(stop)` — Starts at `0`, increments by `1`, stops at `stop - 1`.
- `range(start, stop)` — Starts at `start`, increments by `1`, stops at `stop - 1`.
- `range(start, stop, step)` — Starts at `start`, increments by `step`, stops before reaching or exceeding `stop`.

---

### Sequence Visualizer

Using `list()` forces `range()` to display all generated values at once:

````python
print("range(5):       ", list(range(5)))
print("range(2, 7):    ", list(range(2, 7)))
print("range(2, 10, 2):", list(range(2, 10, 2)))
````

### Iteration Behavior

````python
for number in range(5):
    print(number)
````

### Critical Architectural Observation

> **Half-Open Intervals (`start` to `stop - 1`):** `range(5)` produces **`0, 1, 2, 3, 4`**. The upper bound (`5`) is **never included**.

This non-inclusive upper bound directly mirrors Python's 0-based indexing and slicing conventions (`sequence[start:stop]`).

## 7. Iterating Over Strings

Strings are sequences of individual characters, making them directly iterable with a `for` loop. This is foundational for sequence processing tasks like scanning DNA, RNA, or protein motifs.

### Direct Character Iteration

````python
sequence = "ATGCGTAC"

for nucleotide in sequence:
    print(nucleotide)
````

### Conceptual Comparison: Item vs. Index Iteration

> **Think About It:**
> - In `for nucleotide in sequence:`, what value does `nucleotide` hold on each pass?
> - How is direct iteration different from using `for i in range(len(sequence)):`?

#### Direct Iteration (`for nucleotide in sequence`)
- `nucleotide` binds directly to each **character value** (`'A'`, `'T'`, `'G'`, etc.).
- Clean, pythonic, and readable when positional metadata (index position) isn't needed.

#### Index-Based Iteration (`for i in range(len(sequence))`)
- `i` binds to the **numeric index** (`0, 1, 2, ...`).
- Provides access to both position (`i`) and value via indexing (`sequence[i]`). Useful when inspecting neighboring characters (e.g., codon pairs).

### Index-Based Traversal Example

````python
sequence = "ATGCGTAC"

for i in range(len(sequence)):
    print(f"Index {i}: {sequence[i]}")
````

### Reflection & Analysis

> **Question:** When processing a FASTA file or genomic sequence, when would you prefer direct character iteration versus index-based iteration?

#### My Analysis

1. **Direct Character Iteration is better when:** `You are only concerned with processing or counting individual bases regardless of position (e.g., calculating GC-content percentage, counting total 'A' or 'T' nucleotides, or validating sequence characters). It is cleaner, faster, and more readable.`
2. **Index-Based Iteration is required when:** `You need positional metadata or context relative to neighboring bases—such as identifying mutation sites by position number, inspecting 3-base reading frames (codons: sequence[i:i+3]), checking adjacent bases for restriction digestion sites, or comparing two aligned sequences side-by-side.`

   - Directs the `for` loop to step through every element in `sequence` sequentially.

## 9. Counters

A counter is a variable initialized outside a loop (typically to `0`) and incremented inside the loop whenever a specific condition is met.

Counters allow us to track occurrences, calculate frequency distributions, and perform quality control checks across sequences or datasets.

### Basic Counter Mechanics

````python
count = 0

for number in range(1, 6):
    count += 1

print(f"Total loop iterations counted: {count}")
````

### Biological Example: Targeted Nucleotide Counting

Consider the DNA sequence: **`ATGCGTACGGAATC`**

To count how many times Guanine (`G`) appears:
1. Initialize `g_count = 0` outside the loop.
2. Iterate through every character in the sequence.
3. Use a conditional check (`if nucleotide == "G":`) to increment `g_count` only on matches.

````python
sequence = "ATGCGTACGGAATC"
g_count = 0

for nucleotide in sequence:
    if nucleotide == "G":
        g_count += 1

print(f"Guanine (G) Count: {g_count}")
````

### Challenge: Complete Base Frequency Counter

Extend this counter pattern to compute the exact counts for all four DNA bases (`A`, `T`, `G`, `C`) across the sequence **`ATGCGTACGGAATC`**.

#### Requirements:

2. Traverse the sequence using a single `for` loop.

````python
sequence = "ATGCGTACGGAATC"

# 1. Initialize counters
a_count = 0
t_count = 0
g_count = 0
c_count = 0

# 2. Iterate through sequence and update counters
for nucleotide in sequence:
    if nucleotide == "A":
        a_count += 1
    elif nucleotide == "T":
        t_count += 1
    elif nucleotide == "G":
        g_count += 1
    elif nucleotide == "C":
        c_count += 1

# 3. Print frequency summary report
total_length = len(sequence)

print(f"""
    ======== Base Frequency Counter ========
    Adenine  (A) Count: {a_count}
    Thymine  (T) Count: {t_count}
    Guanine  (G) Count: {g_count}
    Cytosine (C) Count: {c_count}
    ----------------------------------------
    Total Sequence Length: {total_length}
    ========================================
""")
````

## 10. Accumulators

An **accumulator** stores a running total across loop iterations. 

While a counter typically increments by a constant value (`+ 1`) to count *occurrences*, an accumulator adds a dynamic variable payload (`+ value`) on each pass to compute a aggregate sum.

### Basic Accumulator Pattern

````python
measurements = [10, 20, 30, 40]
total = 0

for value in measurements:
    total += value

print(f"Accumulated Total: {total}")
````

### Calculating Sample Statistics (Mean)

Combining an accumulator (`total`) with a length counter (`len(measurements)`) allows us to calculate summary statistics manually.

````python
measurements = [10, 20, 30, 40]
total = 0

for value in measurements:
    total += value

mean = total / len(measurements)

print(f"Total Sum:  {total}")
print(f"Number of samples: {len(measurements)}")
print(f"Sample Mean: {mean}")
````

### Conceptual Connection: Imperative Logic vs. Vectorized Computing

> **Learning Principle:** The manual iteration steps performed above (accumulating values element-by-element to find a sum and mean) are precisely what optimized numerical engines do under the hood.

## 12. Debugging with `print()`

When code produces an incorrect result, guessing why it failed is slow and unreliable. The most effective first step is to **inspect runtime state** by printing variable values at key execution checkpoints.

Instead of treating the program as a black box, we insert strategic `print()` statements to observe how state mutates on every pass through a loop.

### Silent Loop Execution (Hard to Inspect)

````python
measurements = [10, 20, 30]
total = 0

for value in measurements:
    total += value

print("Final total:", total)
````

### Instrumented Loop Execution (State Inspection)

By logging state **before**, **during**, and **after** each iteration, we can verify every intermediate step against our mental model:

````python
measurements = [10, 20, 30]
total = 0

for value in measurements:
    print(f"Before: total = {total}")
    print(f"Adding value: {value}")
    
    total += value
    
    print(f"After:  total = {total}")
    print("-" * 25)

print("Final total:", total)
````

## 14. Logic Errors & Error Classification

A **logic error** occurs when a program runs from start to finish without crashing, but produces an incorrect result. 

Logic errors are the most dangerous class of errors in numerical and scientific computing because Python issues **no warning or traceback**. The code is syntactically valid, but the underlying mathematical or algorithmic logic is flawed.

---

### The Defective Accumulator Example

````python
measurements = [10, 20, 30]
total = 0

for value in measurements:
    total = value  # BUG: Assignment (=) instead of accumulation (+=)

print("Reported total:", total)
````

### Investigation & Trace

Why did the cell above output `30` instead of `60`? Let's trace the mutation of `total` on every pass:

1. **Before loop:** `total = 0`
2. **Iteration 1 (`value = 10`):** `total = 10` (overwrites `0`)
3. **Iteration 2 (`value = 20`):** `total = 20` (overwrites `10`)
4. **Iteration 3 (`value = 30`):** `total = 30` (overwrites `20`)

> **The Root Cause:** Using the assignment operator (`=`) repeatedly overwrites `total` with the current element rather than accumulating it. On each pass, previous work is lost.

---

### Correcting the Logic

````python
measurements = [10, 20, 30]
total = 0

for value in measurements:
    total += value  # Correct: Adds current value to running total

print("Correct total:", total)
````

### Three Common Types of Errors

| Error Type | What Happens? | Python Behavior | Example Cause |
|---|---|---|---|
| **Syntax Error** | Code violates Python language grammar. | Fails immediately *before* execution starts. | Missing colon `:`, unclosed string/bracket. |
| **Runtime Error** | Code is syntactically valid but encounters an impossible operation during execution. | Execution starts, then crashes with an Exception. | Division by zero (`ZeroDivisionError`), invalid index (`IndexError`). |
| **Logic Error** | Code runs to completion without crashing, but performs the wrong algorithm. | Code completes cleanly, but returns incorrect results. | Wrong operator (`=` vs `+=`), off-by-one bounds, flipped Boolean logic. |

````python
sequence = "ATGCGATCGATCGGCTA"

a_count = 0
t_count = 0
g_count = 0
c_count = 0

# 1. Base frequency counts
for nucleotide in sequence:
    if nucleotide == "A":
        a_count += 1
    elif nucleotide == "T":
        t_count += 1
    elif nucleotide == "G":
        g_count += 1
    elif nucleotide == "C":
        c_count += 1

# 2. Total length & GC Proportion calculation
total_bases = len(sequence)

if total_bases > 0:
    gc_content_prop = (g_count + c_count) / total_bases
else:
    gc_content_prop = 0.0

# 3. Formatted Summary
print(f"""
    ================ Summary ================
    Sequence: {sequence} (Length: {total_bases})
    
    Adenine  (A): {a_count}
    Thymine  (T): {t_count}
    Guanine  (G): {g_count}
    Cytosine (C): {c_count}
    
    GC Content Proportion: {gc_content_prop:.2f} ({gc_content_prop * 100:.1f}%)
    =========================================
""")
````

### Loops

* **Why do we need loops?**
  * To automate repetitive tasks across large datasets (such as processing thousands of DNA sequences) without duplicating code, ensuring efficiency and scalability ($O(N)$ execution).
* **What is the difference between `while` and `for`?**
  * A `for` loop iterates over a known, finite sequence or collection. A `while` loop runs continuously as long as a Boolean condition remains `True` (ideal for event-driven or state-dependent stopping conditions).
* **What are the three important components of a `while` loop?**
  1. **Initialization:** Setting up state variables before entering the loop.
  2. **Condition Check:** A Boolean expression in the header evaluated before every pass.
  3. **State Mutation:** Updating variables inside the loop body to progress toward making the condition `False`.
* **What causes an infinite loop?**
  * Failing to mutate the loop condition variable inside the loop body, or setting a condition that can never evaluate to `False`.
* **How does `range()` work?**
  * It generates an immutable sequence of integers following a half-open interval `[start, stop, step)`.
* **What does `range(5)` produce?**
  * `0, 1, 2, 3, 4` (starts at `0` by default, stops before `5`, with a default step of `1`).

---

### Iteration

* **How can I iterate over a string?**
  * By using a direct `for` loop: `for char in sequence:`. Python natively treats strings as ordered sequences of individual characters.
* **What does `in` do?**
  1. **Iterator Syntax:** Used in `for item in collection:` to drive sequence traversal.
  2. **Membership Operator:** Evaluates whether a target substring exists within a sequence, returning `True` or `False`.
* **What is the difference between testing membership with `in` and iterating through a sequence?**
  * Testing membership (`"A" in sequence`) evaluates to a single Boolean result immediately. Iterating (`for base in sequence:`) visits every element sequentially, enabling item-by-item inspection, counting, or conditional checks.

---

### Counters and Accumulators

* **What is a counter?**
  * A variable initialized outside a loop (typically to `0`) and incremented by a fixed step (`count += 1`) inside the loop when a condition is met.
* **What is an accumulator?**
  * A variable initialized outside a loop that aggregates dynamic or varying quantitative values (`total += measurement`) across iterations.
* **What is the difference between them?**
  * A counter tracks **frequency/occurrences** by adding a fixed value (`+1`). An accumulator tracks a **running total/sum** by adding varying values (`+ value`).
* **Can I use a loop to calculate a total?**
  * Yes, by using an accumulator initialized outside the loop (`total = 0`) updated on each pass (`total += value`).
* **Can I use a loop to calculate a mean?**
  * Yes, by combining an accumulator (`total += value`) and a counter (`count += 1`), then dividing `total / count` after loop completion (with a defensive check `if count > 0:`).

### Debugging

* **What is a syntax error?**
  * A violation of Python's formal language rules (e.g., missing colon, unclosed parenthesis) caught *before* code execution starts.
* **What is a runtime error?**
  * An execution crash triggered mid-program when performing an impossible operation (e.g., `ZeroDivisionError`, `IndexError`).
* **What is a logic error?**
  * A silent defect where code runs to completion without crashing, but generates incorrect results due to flawed arithmetic or algorithm design (`=` instead of `+=`).
* **Why can a program run successfully while still being wrong?**
  * Python verifies language mechanics and syntax, not scientific intent. Mathematically valid code can easily compute the wrong biological metric if the algorithm is flawed.
* **How can `print()` help me understand a program?**
  * It exposes hidden internal runtime state **before**, **during**, and **after** loop iterations, revealing where actual execution diverges from your expected mental model.
* **What does it mean to trace a program?**
  * Stepping through code line-by-line manually to record variable state mutations and conditional outputs at every step.
