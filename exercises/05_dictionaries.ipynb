## 2. Dictionaries vs. Lists

A **list** uses positional indices:

````python
students = ["Amina", "Bello", "Chika"]
grades = [87, 72, 91]

print(students[0])
print(grades[0])
````

Here, `0` means the first position.

A **dictionary** represents the relationship directly:

````python
grades = {
    "Amina": 87,
    "Bello": 72,
    "Chika": 91
}

print(grades["Amina"])
````

### Comparison

```text
List Approach:
students[0] → first position (must track separate lists in parallel)

Dictionary Approach:
grades["Amina"] → value directly associated with "Amina"


This is one reason dictionaries are useful for real-world data: I can retrieve information using a meaningful identifier without needing to track positional indices across separate structures.
```

### Investigate

````python
# Adding to list version
students.append("David")
grades_list = [87, 72, 91]
grades_list.append(84)

# Look up Bello's grade using lists
bello_index = students.index("Bello")
print("List lookup for Bello's grade:", grades_list[bello_index])

# Adding to dictionary version
grades_dict = {
    "Amina": 87,
    "Bello": 72,
    "Chika": 91
}
grades_dict["David"] = 84

# Look up Bello's grade using dictionary
print("Dictionary lookup for Bello's grade:", grades_dict["Bello"])
````

### Observation & Structural Analysis

#### Diagnostic Comparison
**Which version makes it easier to answer: *"What is Bello's grade?"***

The **dictionary version** (`grades_dict["Bello"]`) is vastly superior to tracking parallel lists.

---

#### Core Advantages of Dictionaries

1. **Direct Label Access:**  
   Retrieves Bello's grade directly via `grades_dict["Bello"]` without needing to know or track where Bello is stored in memory.

2. **Elimination of Parallel Index Overhead:**  
   When using parallel lists (`students` and `grades_list`), querying a value requires locating `"Bello"`'s positional index using `students.index("Bello")`, then passing that index into `grades_list[idx]`.

3. **Structural Safety:**  
   If parallel lists undergo insertion, deletion, or sorting, their positional alignment breaks, leading to corrupted data relationships. Dictionaries explicitly couple keys to values inside a single unified structure.

---

### Key Takeaway

> Assigning a value to an existing key **updates** its value in-place. Assigning a value to a non-existent key **inserts** a brand-new key-value pair.

## 4. Accessing Data Safely

A direct lookup assumes that the key exists:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax"
}

print(sample["species"])
````

But attempting to access a missing key using standard bracket notation causes an error:

````python
print(sample["host"])
````

The error is a `KeyError`. Python is telling me that `"host"` is not a key in the dictionary.

### Checking with `in`
Before accessing a key, I can check whether it exists using the `in` containment operator:

````python
if "host" in sample:
    print(sample["host"])
else:
    print("Host information is not available")

print("species" in sample)
print("host" in sample)
````

The `in` operator checks **dictionary keys**, returning `True` if the key exists and `False` otherwise.

---

### Safe Retrieval using `.get()`

Another option for safe key access is the `.get()` method:

````python
print(sample.get("host"))
````

If the key does not exist, `.get()` returns `None` instead of raising a `KeyError`.

I can also provide a custom fallback default value as a second argument:

````python
print(sample.get("host", "unknown"))
````

### Conceptual Comparison

```text
sample["host"]
    ↓
Direct Access: I expect the key to exist (raises KeyError if missing).

"host" in sample
    ↓
Membership Check: I want to explicitly check whether the key exists before acting.

sample.get("host", "default")
    ↓
Safe Retrieval: The key may be missing, and I want a safe fallback value without crashing.

### Investigate
Let's test `.get()` before and after adding `"host"` to the dictionary:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax"
}

# 1. Before adding "host"
print("Before:", sample.get("host", "not recorded"))

# 2. Add "host" to the dictionary
sample["host"] = "cattle"

# 3. Run the exact same line again
print("After:", sample.get("host", "not recorded"))
````

### Observation & Diagnostic Explanation

#### Observation
* **Before key insertion:** `sample.get("host", "not recorded")` evaluates to `"not recorded"`.
* **After key insertion:** `sample.get("host", "not recorded")` evaluates to `"cattle"`.

#### Diagnostic Explanation
* **Missing Key Guard:** On the first execution, Python searches the hash table inside `sample` for the key `"host"`. Finding no match, `.get()` suppresses the `KeyError` and gracefully returns the fallback parameter (`"not recorded"`).
* **Existing Key Access:** Once `sample["host"] = "cattle"` executes, `"host"` exists in the dictionary. On the second execution, `.get()` retrieves the stored value (`"cattle"`) directly and ignores the fallback argument.

---

### Key Concept: `.get()` Method Mechanics

```python
# Syntax Pattern:
dictionary.get(key, default_value)

- **Direct Bracket Access (`dict[key]`):** Use when a missing key represents an unexpected pipeline error that should stop program execution.

- **Safe `.get()` Access (`dict.get(key, default)`):** Use when processing incomplete datasets or optional biological metadata fields where missing attributes are expected and need fallback values.

## 5. Removing Data

There are two primary ways to remove dictionary entries: `del` and `.pop()`.

### Using `del`

The `del` statement removes a key-value pair directly from the dictionary:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax",
    "host": "cattle"
}

del sample["host"]

print(sample)
````

`del` permanently removes the specified key and its associated value.

---

### Using `.pop()`

The `.pop()` method also removes a key-value pair, but it returns the removed value so it can be captured or stored in a variable:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax",
    "host": "cattle"
}

removed_host = sample.pop("host")

print("Removed:", removed_host)
print(sample)
````

Unlike `del`, `.pop()` allows me to work with the value after deleting it from the dictionary structure.

---

### Predict

What do I expect this to produce?

```python
result = sample.pop("missing")

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax"
}

result = sample.pop("missing")
````

### Investigate
Now let's test whether providing a second argument works when popping a missing key:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax"
}

result = sample.pop("missing", "not found")

print("Result:", result)
print("Dictionary status:", sample)
````

### Observation & Diagnostic Explanation

#### Observation
1. `sample.pop("missing")` raises a `KeyError: 'missing'`.
2. `sample.pop("missing", "not found")` returns `"not found"` and continues execution safely.

#### Diagnostic Explanation
* **Unprotected Deletion (`.pop(key)`):** `.pop()` attempts to remove the key-value pair *and* return the removed value. If the key does not exist in the hash table, Python cannot return a value, causing an immediate `KeyError`.
* **Defensive Deletion (`.pop(key, default)`):** Supplying a second argument acts as a fallback handler. If the key is found, Python removes it and returns its associated value; if the key is missing, Python returns the default argument (`"not found"`) without throwing an exception.

---

### Key Concept: Dictionary Deletion Patterns

| Method | Syntax | Missing Key Behavior | Primary Use Case |
| :--- | :--- | :--- | :--- |
| **`del` Statement** | `del dict[key]` | Raises `KeyError` | Direct deletion when the key is guaranteed to exist. |
| **Unprotected `.pop()`** | `val = dict.pop(key)` | Raises `KeyError` | Deletes entry AND captures its value for downstream processing. |
| **Defensive `.pop()`** | `val = dict.pop(key, default)` | Returns `default` value | Safe removal from optional or incomplete datasets without breaking loops. |

## 6. Inspecting and Iterating Through Dictionaries

Python provides several built-in methods and functions to inspect dictionary structures:

````python
sample = {
    "id": "AAT001",
    "species": "Trypanosoma vivax",
    "host": "cattle",
    "assay": "ELISA"
}

print("Length:", len(sample))
print("Keys:", sample.keys())
print("Values:", sample.values())
print("Items:", sample.items())
````

These methods represent:

* **`len(dict)`** → total number of key-value pairs.
* **`.keys()`** → a view object containing all dictionary keys.
* **`.values()`** → a view object containing all dictionary values.
* **`.items()`** → a view object containing key-value tuples `(key, value)`.

---

### Iterating Over Keys

By default, looping directly over a dictionary iterates over its **keys**:

````python
for key in sample:
    print(key)
````

This is equivalent in functionality to explicitly calling `.keys()`:

````python
for key in sample.keys():
    print(key)
````

---

### Iterating Over Values

To loop exclusively over the values stored in a dictionary, use `.values()`:

````python
for value in sample.values():
    print(value)
````

---

### Iterating Over Key-Value Pairs

To access both the key and its corresponding value in each iteration step, use `.items()` combined with sequence unpacking:

````python
for key, value in sample.items():
    print(key, ":", value)
````

The `.items()` pattern is particularly powerful when formatting summary reports or transferring metadata into downstream processing pipelines.

---

### Observation & Iteration Mechanics

#### Observation
The direct loop `for x in sample:` printed only the keys (`"id"`, `"species"`, `"host"`, `"assay"`), leaving the associated values (`"AAT001"`, `"Trypanosoma vivax"`, etc.) unrendered.

#### Diagnostic Explanation
* **Default Iteration Rule:** In Python, iterating directly over a dictionary (e.g., `for item in my_dict:`) iterates exclusively over its **keys**. The hash table yields keys sequentially to optimize loop speed and memory overhead.
* **Accessing Associated Values:** To access values during iteration, you must either:
  1. Perform an explicit lookup inside the loop body using bracket notation (`sample[key]`).
  2. Use the **`.items()`** view method to yield key-value tuples directly and unpack them into loop variables (`for key, value in sample.items():`).

---

### Key Concept: Dictionary View Methods

| View Method | Yielded Element | Canonical Loop Syntax | Primary Use Case |
| :--- | :--- | :--- | :--- |
| **`dict` (Direct)** | Keys only | `for k in sample:` | Quick key validation, field checks. |
| **`.keys()`** | Keys view | `for k in sample.keys():` | Explicit readability when working solely with keys. |
| **`.values()`** | Values view | `for v in sample.values():` | Statistical metrics (e.g., mean concentration across samples). |
| **`.items()`** | `(key, value)` tuples | `for k, v in sample.items():` | **Best practice** for multi-attribute reporting and transformations. |

## 9. Dictionaries as Lookup Tables

A **lookup table** connects one piece of information directly to another.

````python
species_codes = {
    "Tco": "Trypanosoma congolense",
    "Tv": "Trypanosoma vivax",
    "Te": "Trypanosoma evansi"
}

print(species_codes["Tv"])
````

### Replacing Conditional Chains

Instead of writing verbose `if-elif-else` control flow:

```python
if code == "Tco":
    full_name = "Trypanosoma congolense"
elif code == "Tv":
    full_name = "Trypanosoma vivax"
elif code == "Te":
    full_name = "Trypanosoma evansi"

a dictionary represents the relationship directly, reducing boilerplate code and making your logic cleaner and faster.

### Conceptual Flow

code ("Tv")
    ↓
species_codes lookup
    ↓
"Trypanosoma vivax"

This is one of the most common practical design patterns for dictionaries in bioinformatics and data analysis.

````python
# Practical Example: Mapping a list of sample codes to full scientific names
sample_codes = ["Tv", "Tco", "Tv", "Te", "Tv"]

for code in sample_codes:
    full_name = species_codes.get(code, "Unknown Species")
    print(f"Code: {code:<4} -> Species: {full_name}")
````

---

### Why Use a Lookup Table?

1. **Efficiency:** Dictionary key lookups run in **constant time ($O(1)$)** regardless of how many key-value pairs are stored, whereas long `if-elif` chains evaluate sequentially ($O(N)$).
2. **Maintainability:** Adding a new mapping requires adding one entry to the dictionary, rather than writing a new `elif` block.
3. **Separation of Data and Logic:** The dataset/mappings are stored independently from the code that processes them.

## 10. Dictionaries + Loops: Counting Frequencies

One of the most important dictionary patterns in computer science and data analysis is **counting how often items occur**.

Suppose I have a list of identified genes from a sequencing assay:

```python
genes = [
    "ISG75",
    "HSP70",
    "ISG75",
    "p64",
    "HSP70",
    "ISG75"
]
```

I want to process this list to generate a frequency summary dictionary:

```python
{
    "ISG75": 3,
    "HSP70": 2,
    "p64": 1
}


````python
genes = [
    "ISG75",
    "HSP70",
    "ISG75",
    "p64",
    "HSP70",
    "ISG75"
]

counts = {}

for gene in genes:
    if gene in counts:
        counts[gene] += 1
    else:
        counts[gene] = 1

print(counts)
````

### Understanding the Algorithm

The decision logic inside the loop answers one question during each iteration:

**"Have I seen this item before?"**

```text
       Have I seen this item before?
               /          \
            YES            NO
             ↓              ↓
      Increase count   Create entry
       (counts[x] += 1) (counts[x] = 1)

### Tracing the First Few Iterations
Tracing `genes = ["ISG75", "HSP70", "ISG75"]`:

1. Initial state: `{}`

2. First item (`"ISG75"`): Not in `counts` → set `counts["ISG75"] = 1`

    - State: `{"ISG75": 1}`

2. Second item (`"HSP70"`): Not in `counts` → set `counts["HSP70"] = 1`

    - State: `{"ISG75": 1, "HSP70": 1}`

3. Third item (`"ISG75"`): Found in `counts` → increment `counts["ISG75"] += 1`

    - State: `{"ISG75": 2, "HSP70": 1}`

---

### Alternative Implementation: Idiomatic `.get()` Pattern

Instead of using an explicit `if-else` block, you can accomplish frequency counting in a single line inside the loop using `.get()` with a default value of `0`:

````python
counts = {}

for gene in genes:
    counts[gene] = counts.get(gene, 0) + 1

print("Frequency count via .get():", counts)
````

### Conceptual Synthesis: The Frequency Counter Pattern

The **Frequency Counter Pattern** is a foundational algorithmic technique in computational data pipelines. By using a dictionary to map discrete items (keys) to their occurrence counts (values), you can aggregate large streams of categorical data in a single pass ($\mathcal{O}(N)$ time complexity).

```python
# Canonical Frequency Counter Pattern
counts = {}
for item in dataset:
    counts[item] = counts.get(item, 0) + 1

### Broad Domain Applications
- **Bioinformatics:** Quantifying nucleotide/codon frequencies, k-mer counting, and gene expression distribution across sequencing reads.

- **Ecology & Epidemiology:** Aggregating species abundance counts, tracking pathogen strain frequencies, and monitoring regional disease outbreak reports.

- **Text Mining & Natural Language Processing (NLP):** Computing word frequencies, constructing document vocabularies, and computing TF-IDF weights.

- **Data Cleansing & Validation:** Spotting categorical anomalies, measuring missing/null field rates, and auditing raw data distributions prior to downstream modeling.

## 11. From Lists to Dictionaries

In `04_lists.ipynb`, we used parallel lists combined with the built-in `zip()` function:

````python
sample_ids = ["A01", "A02", "A03"]
concentrations = [12.4, 8.7, 15.2]

pairs = list(zip(sample_ids, concentrations))

print(pairs)
````

The resulting pairs represent direct relationships:

```text
A01 → 12.4
A02 → 8.7
A03 → 15.2

This relationship can be converted directly into a dictionary using `dict()` combined with `zip():`

````python
samples = dict(zip(sample_ids, concentrations))

print(samples)
````

Now, instead of manually tracking matching indices, I can retrieve a concentration directly using its sample ID:

````python
print(samples["A02"])
````

### Conceptual Architecture: The Data-Coupling Pipeline

Moving from disconnected parallel sequences to key-value mappings follows a structured, four-stage evolutionary pipeline:

```text
1. Parallel Lists       ("A01", "A02")  and  (12.4, 8.7)
       ↓  [zip()]
2. Zipped Iterative     [("A01", 12.4), ("A02", 8.7)]
       ↓  [dict()]
3. Dictionary Container  {"A01": 12.4, "A02": 8.7}
       ↓  [Direct Access]
4. Constant-Time Lookup  samples["A02"] → 8.7

### Stage-by-Stage Breakdown
1. **Parallel Lists (Uncoupled Storage):** Data exists in separate memory allocations. Structural integrity depends entirely on matching index positions (`sample_ids[1]` maps to `concentrations[1]`). Risk of misalignment is high if either list is modified.
2. **Zipped Collections (`zip()`):** Pairs items positionally into immutable tuples (`(key, value)`). This locks the relational context together, preventing accidental desynchronization.
3. **Dictionary Constructor (`dict()`):** Consumes key-value tuple pairs and builds an internal hash table. Each unique key is hashed to generate a direct pointer address for its mapped value.
4. **Dictionary Lookup ($\mathcal{O}(1)$ Access):** Querying `samples["A02"]` hashes `"A02"` instantly, bypassing sequential searches and retrieving `8.7` in constant time regardless of dataset size.

## 13. A Brief Look at Dictionary Comprehensions

## 14. Dictionary Keys and Data Types

Dictionary keys need to be values that Python can use reliably as identifiers.

Common valid key examples include strings:

````python
sample = {
    "AAT001": "positive",
    "AAT002": "negative"
}

print(sample)
````

And single characters or short string labels:

Strings, integers, floats, and tuples can all serve as dictionary keys.

**Lists, however, cannot be used as keys:**

```python

    ["A", "T"]: "DNA"
}

Let's test this in code to see what error Python raises:

````python
data = {
    ["A", "T"]: "DNA"
}
````

### Observation & Explanation
**Observation:**
Attempting to use a list `["A", "T"]` as a dictionary key causes a `TypeError: unhashable type: 'list'`.

**Explanation:**
Because dictionaries use internal hash tables to locate keys instantly ($O(1)$ time complexity), a key's identity must remain fixed and unchanging over time.
- Immutable types (strings, numbers, tuples) cannot be altered after creation. Python can compute a fixed hash code for them, making them valid keys.
- Mutable types (lists, dictionaries, sets) can be changed in place (e.g., via `.append()`). If a list were allowed as a key and then modified, its hash value would change, breaking the lookup mechanism inside the dictionary.

### Key Takeaway
Not every Python object can be used as a dictionary key.

For almost all data processing tasks, using strings, integers, or tuples as keys will cover your needs cleanly and safely.

### Debugging Exercise 1 — Missing Key

Consider the following snippet:

```python
sample = {
    "id": "AAT001",
    "species": "T. vivax"
}

print(sample["host"])
```

### Prediction
Directly indexing a non-existent key using bracket notation (`sample["host"]`) will cause the Python interpreter to halt execution and raise a `KeyError: 'host'`.

````python
# Run the broken code to observe the exact trace
sample = {
    "id": "AAT001",
    "species": "T. vivax"
}

print(sample["host"])
````

### Error Identification & Root Cause Analysis
-   **Error Raised:** `KeyError`
-   **Root Cause:** Standard bracket access (`dict[key]`) requires that the requested key exists within the dictionary's hash table. When Python evaluates `sample["host"]`, it searches the key namespace, finds no matching key named `"host"`, and immediately halts with a `KeyError`.

### Solutions: Defensive Access Patterns
To prevent program crashes when inspecting potentially missing fields, we can use two defensive strategies:

````python
sample = {
    "id": "AAT001",
    "species": "T. vivax"
}

# Fix Option 1: Pre-check key existence using the 'in' operator
if "host" in sample:
    print("Host (via in):", sample["host"])
else:
    print("Host (via in): 'host' key is not present in record.")

# Fix Option 2: Use .get() with a fallback default value
host_info = sample.get("host", "Unknown Host")
print("Host (via .get()):", host_info)
````

### Why the Original Code Failed

When using direct bracket notation (`dict[key]`), Python assumes you are confident the key exists. If it does not, Python raises an unhandled exception rather than returning a blank or `None` value silently. 

* **Use `dict[key]`** when the key's presence is mandatory, and a missing key represents a critical data bug.
* **Use `dict.get(key, default)`** when dealing with optional, missing, or sparse biological metadata fields.

### Debugging Exercise 2 — Incorrect Iteration

Consider the following loop attempting to inspect dictionary contents:

```python
sample = {
    "id": "AAT001",
    "species": "T. vivax",
    "result": "positive"
}

for key, value in sample:
    print(key, value)
```
### Prediction
Attempting to unpack sample directly into two variables (key, value) will fail because iterating directly over a dictionary yields only individual keys (strings), not pairs.

````python
# Run the broken code to observe the exact trace
sample = {
    "id": "AAT001",
    "species": "T. vivax",
    "result": "positive"
}

for key, value in sample:
    print(key, value)
````

### Error Identification & Root Cause Analysis
- **Error Raised:** `ValueError`
- **Root Cause:** By default, iterating directly over a dictionary (`for x in sample`) yields its keys one by one. On the first iteration, Python yields the string key `"id"`. The loop then attempts tuple unpacking: `key`, `value = "id"`. Because `"id"` is a 2-character string, Python attempts to unpack `'i'` into key and `'d'` into value. When it encounters `"species"`, it has 7 characters and raises `ValueError: too many values to unpack (expected 2)`.

### Corrected Implementation
To iterate over both keys and values simultaneously, call the `.items()` view method:

````python
sample = {
    "id": "AAT001",
    "species": "T. vivax",
    "result": "positive"
}

# Corrected iteration structure using .items()
for key, value in sample.items():
    print(f"{key:<10} : {value}")
````

### Why `.items()` Works (Under the Hood)

1. **View Object of Tuples:** The `.items()` method returns a dynamic dictionary view containing 2-element tuples:
   `[("id", "AAT001"), ("species", "T. vivax"), ("result", "positive")]`

2. **Sequence Unpacking:** On each loop iteration, Python retrieves one tuple `(k, v)` and automatically assigns:
   * Index `0` (the key) $\rightarrow$ `key`
   * Index `1` (the value) $\rightarrow$ `value`

This makes `for key, value in sample.items():` clean, readable, and structurally safe.

### Debugging Exercise 3 — Counter Logic

Consider the following program attempting to count gene frequencies:

```python
genes = ["ISG75", "HSP70", "ISG75"]

counts = {}

for gene in genes:
    counts[gene] = 1

print(counts)
```

### Prediction
The output will be `{'ISG75': 1, 'HSP70': 1}` instead of the expected `{'ISG75': 2, 'HSP70': 1}`.

````python
# Run the logic bug code to inspect output
genes = ["ISG75", "HSP70", "ISG75"]

counts = {}

for gene in genes:
    counts[gene] = 1

print("Incorrect Result:", counts)
````

### Error Analysis: Syntax vs. Logic Errors

* **Type of Error:** **Logic Error** (Semantic Error)
* **Why `ISG75` does not end with 2:**  
  The assignment statement `counts[gene] = 1` **re-initializes** the value to `1` every single time the loop sees `"ISG75"`. 
  * On iteration 1 (`"ISG75"`): `counts["ISG75"]` becomes `1`.
  * On iteration 2 (`"HSP70"`): `counts["HSP70"]` becomes `1`.
  * On iteration 3 (`"ISG75"`): `counts["ISG75"]` is overwritten with `1` again, discarding the fact that it was seen previously.

### Key Operational Difference

| Assignment Statement | Action | Behavior on Existing Key |
| :--- | :--- | :--- |
| `counts[gene] = 1` | Direct Assignment | **Overwrites** existing value with hardcoded `1` |
| `counts[gene] += 1` | Augmented Addition | **Increments** stored integer value by `1` |

`counts[gene] += 1` is shorthand for `counts[gene] = counts[gene] + 1`. This requires reading the previous count before adding to it.

````python
# Corrected Frequency Counter Logic
genes = ["ISG75", "HSP70", "ISG75"]

counts = {}

for gene in genes:
    if gene in counts:
        counts[gene] += 1  # Increment existing tally
    else:
        counts[gene] = 1   # Initialize new key

print("Corrected Result:", counts)
````

### Debugging Exercise 4 — Key Type

Consider attempting to use a list as a dictionary key:

```python
data = {
    ["A", "T", "G", "C"]: "DNA"
}

````python
# Run the broken code to inspect the runtime error
data = {
    ["A", "T", "G", "C"]: "DNA"
}
````

### Error Identification & Root Cause Analysis
- **Error Raised:** `TypeError: unhashable type: 'list'`
- **Root Cause:** Dictionaries rely on an internal memory structure called a hash table to achieve instant $O(1)$ lookups. To place an object in a hash table, Python computes a fixed "fingerprint" (hash value) from the key. Because lists are mutable (their contents can be changed at any time with `.append()` or `[]` assignment), their internal fingerprint would change if modified. If a key's fingerprint changes, Python can no longer find where the value was stored in memory.

### Corrected Implementation: Replacing List with Tuple
To represent a sequence of elements as a dictionary key, convert the mutable list `[...]` into an immutable tuple `(...)`.

````python
# Fixed dictionary using a tuple key
data = {
    ("A", "T", "G", "C"): "DNA",
    ("A", "U", "G", "C"): "RNA"
}

# Lookup works as expected
bases = ("A", "T", "G", "C")
print("Lookup Result:", data[bases])
````

### Summary & Core Takeaway

A dictionary key **must** be a data type that remains completely unchanged throughout the execution of the program. 

* **Valid Keys (Immutable):** Strings (`"AAT001"`), Integers (`42`), Floats (`3.14`), Tuples (`("A", "T")`).
* **Invalid Keys (Mutable):** Lists (`["A", "T"]`), Dictionaries (`{"a": 1}`), Sets (`{"A", "T"}`).

### 18.1 Core Dictionary Concepts

* **What is a dictionary?**  
  A mutable, key-indexed data structure that maps unique keys to values, enabling $O(1)$ lookups via meaningful labels instead of integer positions (e.g., `sample["species"]`).
* **Keys vs. Values:**  
  * **Key:** A unique, immutable lookup identifier (e.g., `"species"`).
  * **Value:** The data stored under that identifier (e.g., `"T. vivax"`).
* **When to use a Dictionary vs. a List:**  
  Use **lists** when sequence order matters. Use **dictionaries** when establishing named relationships (e.g., `sample_id -> concentration`).

### 18.2 Accessing and Modifying Data

* **Direct Access:** `sample["species"]` extracts a value. Raises a `KeyError` if the key does not exist.
* **Safe Checking (`in` operator):** `'host' in sample` evaluates to `True` or `False` without throwing an error.
* **Fallback Access (`.get()`):** `sample.get("host", "Unknown")` returns the value if present, or a fallback default without crashing.
* **Mutating Entries:** `sample["host"] = "cattle"` assigns a value. It creates a new entry if the key is missing or updates the value if it already exists.
* **Deletion:** `del sample["host"]` removes an entry; `sample.pop("host")` removes and returns the value.

### 18.3 Iteration & Frequency Counting

* **Default Loop:** `for k in sample:` iterates exclusively over dictionary **keys**.
* **View Methods:** `.keys()` yields keys, `.values()` yields values, and `.items()` yields `(key, value)` tuples.
* **Why `.items()` is Essential:** It enables simultaneous sequence unpacking of keys and values:
  ```python
  for key, value in sample.items():
      print(key, value)

### 18.4 Advanced Structures: Lists, Zip, and Nesting

* **Combining Parallel Lists (`zip`):**  
  `dict(zip(ids, concs))` merges two positional lists into explicit key-value pairs (`{"S001": 12.5}`). Truncates silently if list lengths differ.
* **Nested Dictionaries:**  
  Allows multi-attribute tracking for structured datasets (e.g., biological samples or JSON API outputs).
* **Chained Access:** Accessing deep fields uses sequential brackets:
  ```python
  samples["AAT001"]["species"]  # Outer lookup ("AAT001") -> Inner lookup ("species")
* **Dictionary Keys Must Be Immutable (Hashable):**
    Strings, integers, and tuples are valid keys; mutable objects like lists raise a `TypeError`.
* **Dictionary Comprehensions:**
    Construct dictionaries dynamically: `{gene: 0 for gene in genes}`.

### 18.5 Debugging Cheat Sheet

| Exception / Issue | Common Root Cause | Prevention / Solution |
| :--- | :--- | :--- |
| `KeyError` | Accessing a non-existent key via `dict[key]` | Use `"key" in dict` check or `dict.get(key, default)` |
| `ValueError` | Unpacking a dictionary loop without `.items()` | Iterate using `for key, value in dict.items():` |
| **Logic Error** | Using `counts[item] = 1` inside a loop | Use `counts[item] += 1` to increment instead of overwriting |
| `TypeError: unhashable` | Attempting to use a `list` as a dictionary key | Convert the key sequence to an immutable `tuple` |
