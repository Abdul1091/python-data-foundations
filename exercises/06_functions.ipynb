## 1. Why Functions?

Before creating a function, let's observe the core issue that functions solve: **code duplication**.

Suppose we have optical density ($OD_{450}$) antibody assay measurements for two experimental 

### The Problem: Manual Duplication

````python
# Sample A: Optical density readings
sample_a = [2.1, 3.4, 1.8, 4.2]

total = 0
for value in sample_a:
    total += value

mean_a = total / len(sample_a)
print(f"Sample A Mean: {mean_a:.2f}")

# Sample B: Optical density readings
sample_b = [5.2, 4.7, 6.1, 5.8]

total = 0
for value in sample_b:
    total += value

mean_b = total / len(sample_b)
print(f"Sample B Mean: {mean_b:.2f}")
````

### Investigation & Architectural Trade-offs

If we scaled this pipeline to analyze **20 or 100 sample plates**, copying and pasting this logic causes three major structural vulnerabilities:

1. **Maintenance Bottleneck:** If we change how missing values or zero readings are handled, we must edit every single block individually.
2. **Variable Collisions:** Reusing accumulator variables like `total` across script blocks increases the risk of stale values leaking into subsequent calculations.
3. **Readability & Noise:** The underlying scientific intent (calculating an average) gets buried inside mechanical loop boilerplate.

### The Functional Solution

A function encapsulates the calculation logic **once** under a single reusable label.

$$\text{Input Data (Argument)} \longrightarrow \boxed{\text{Function Logic}} \longrightarrow \text{Output Value (Return)}$$

````python
# Encapsulating the calculation into a reusable function
def calculate_mean(measurements):
    """Calculates the mean value of a list of numbers."""
    total = 0
    for value in measurements:
        total += value
    return total / len(measurements)

# Executing the function across different datasets
mean_a = calculate_mean(sample_a)
mean_b = calculate_mean(sample_b)

print(f"Sample A Mean (Functional): {mean_a:.2f}")
print(f"Sample B Mean (Functional): {mean_b:.2f}")
````

## 2. Creating a Function

Defining a function creates a blueprint; it does not execute the code immediately. Python stores the instructions under the function's name until it is explicitly called.

### Key Mechanics:
1. **`def` Keyword:** Informs Python that a function definition is starting.
2. **Parameters (`values`):** Temporary placeholders inside parentheses that accept data passed into the function.
3. **Indented Body:** The block of code that runs whenever the function is called.
4. **`return` Statement:** Sends the calculated value back to where the function was called and immediately exits the function.

````python
# 1. DEFINE the logic once
def calculate_mean(values):
    """Calculates the mean value from an iterable of numerical values."""
    total = 0
    for value in values:
        total += value
    return total / len(values)

# Note: Running this cell alone produces no printed output because the function is only defined.
````

### Invoking (Calling) the Function

Once defined, we invoke the function by passing different datasets into its parameter list.

$$\text{Define Logic} \longrightarrow \text{Provide Data} \longrightarrow \text{Reuse Execution}$$

````python
# 2. PROVIDE DATA & REUSE LOGIC

# Pass a named variable
sample_a = [2.1, 3.4, 1.8, 4.2]
mean_a = calculate_mean(sample_a)
print(f"Sample A Mean: {mean_a:.2f}")

# Pass a new variable directly
sample_b = [5.2, 4.7, 6.1, 5.8]
print(f"Sample B Mean: {calculate_mean(sample_b):.2f}")
````

## 3. Parameters and Arguments

Understanding the distinction between parameters and arguments is essential for reading function signatures and debugging inputs.

### Key Distinctions

* **Parameter:** The variable placeholder listed inside the function definition (`def`).
* **Argument:** The actual concrete value or variable passed into the function when invoking it.

$$\text{Parameter (Placeholder)} \xleftarrow{\text{Binds Data}} \text{Argument (Actual Value)}$$

````python
# 'values' is the PARAMETER
def calculate_mean(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)

sample_a = [2.1, 3.4, 1.8, 4.2]

# 'sample_a' is the ARGUMENT supplied to the function
mean_a = calculate_mean(sample_a)
print(f"Sample A Mean: {mean_a:.2f}")
````

### Multiple Parameters & Positional Order

Functions can accept multiple parameters separated by commas. By default, arguments are **positional**: Python maps the first argument to the first parameter, the second argument to the second parameter, and so on.

```python
def calculate_difference(a, b):
    return a - b

## 4. Returning Values

One of the most important concepts in function design is understanding the operational difference between `print()` and `return`.

### Mental Model Breakdown

* **`print()` $\longrightarrow$ Displays something on screen.** It produces a visual side-effect for human inspection, but sends **nothing** back to the program.
* **`return` $\longrightarrow$ Handing a value back to the program.** It hands the result back so other lines, variables, loops, or conditional statements can use it.

```text
  print()  ---> [ Standard Output (Console Screen) ] ---> Cannot be reused in code
  return   ---> [ Back to Variable / Expression ]   ---> Reusable in downstream operations

### The Trap: Printing Instead of Returning

Observe what happens when a function prints its result instead of returning it:

````python
# INCORRECT DESIGN: Using print() inside the function
def calculate_mean_display(values):
    total = 0
    for value in values:
        total += value
    print(total / len(values))  # Displays value, but returns None implicitly

# Calling the printing function
result = calculate_mean_display([2, 4, 6])

# Inspecting what 'result' actually captured
print("Captured result variable:", result)
````

#### Explanation & Inspection:
1. The first output (`4.0`) came directly from `print()` inside `calculate_mean_display`.
2. Because the function lacked an explicit `return` statement, Python automatically returned `None`.
3. Consequently, `result` holds `None` instead of the calculated value `4.0`.

### Correct Design: Using `return`

Now let's modify the function to use `return`:

````python
# CORRECT DESIGN: Returning the value
def calculate_mean(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)

# Capturing the returned value into a variable
mean_val = calculate_mean([2, 4, 6])
print("Captured mean_val variable:", mean_val)
````

### Pipelining Returned Values into Downstream Logic

When a function returns a value, we can pipe that output directly into conditional branching, downstream math operations, or store it in structured data stores across various scientific domains.

````python
# Scientific Example 1: DNA Sequence GC Content Analysis
def calculate_gc_content(sequence):
    """Returns the percentage of G and C bases in a DNA sequence."""
    gc_count = 0
    for base in sequence.upper():
        if base in ['G', 'C']:
            gc_count += 1
    return (gc_count / len(sequence)) * 100

dna_seq = "ATGCGATCGATCGATC"
gc_pct = calculate_gc_content(dna_seq)

# Downstream decision based on returned value
if gc_pct > 50:
    print(f"High GC Content ({gc_pct:.1f}%): Suitable for high-temperature PCR priming.")
else:
    print(f"Standard/Low GC Content ({gc_pct:.1f}%).")


# Scientific Example 2: Environmental Temperature Monitoring
def celsius_to_kelvin(temp_c):
    """Converts a temperature measurement from Celsius to Kelvin."""
    return temp_c + 273.15

incubation_temp_c = 37.0
incubation_temp_k = celsius_to_kelvin(incubation_temp_c)

print(f"Incubator Setting: {incubation_temp_c}°C -> {incubation_temp_k} K")
````

## 5. Functions with Lists, Loops, and Conditionals

Functions reach their full utility when they serve as containers for multi-step logic—combining iteration (`for`), conditional branching (`if`), and state tracking (accumulators/counters).

### Building Blocks of a Data-Processing Function

When combining constructs, trace how data moves through the four structural elements:
1. **Input Data (List):** Passed as an argument into the function parameter.
2. **Iteration (`for` loop):** Traverses the sequence element-by-element.
3. **Decision Logic (`if` statement):** Filters or checks each element against a condition.
4. **Output (`return`):** Hands back the accumulated or processed result after the loop finishes.

````python
# Encapsulating list iteration, conditional filtering, and counting
def count_positive(results):
    """Counts occurrences of 'positive' strings in a list of diagnostic results."""
    count = 0
    for result in results:
        if result == "positive":
            count += 1
    return count

# Run 1: Diagnostic assay batch 1
batch_1 = ["positive", "negative", "positive", "positive"]
positive_count_1 = count_positive(batch_1)

print(f"Batch 1 Positive Count: {positive_count_1}")
````

### Experimentation & Reusability

To test reusability, pass a completely different dataset into the exact same function without changing a single line of the function's internal logic.

````python
# Run 2: Diagnostic assay batch 2
batch_2 = [
    "negative",
    "negative",
    "positive",
    "negative",
    "positive"
]

# Predicting outcome: 2 positives out of 5 samples
positive_count_2 = count_positive(batch_2)
print(f"Batch 2 Positive Count: {positive_count_2}")
````

### Expanding Function Logic (Scientific Contexts)

Let's modify this pattern to solve more complex scientific problems—such as filtering numerical thresholds (e.g., optical density cutoffs or protein concentration flags) and analyzing general life science datasets.

````python
# Life Science Application 1: Counting samples above an absorbance threshold
def count_high_absorbance(od_readings, cutoff=1.0):
    """Counts how many optical density readings exceed a specified cutoff value."""
    above_cutoff = 0
    for reading in od_readings:
        if reading > cutoff:
            above_cutoff += 1
    return above_cutoff

serum_samples_od = [0.24, 1.15, 0.88, 1.42, 0.05, 1.05]
reactive_count = count_high_absorbance(serum_samples_od, cutoff=1.0)

print(f"Total samples tested: {len(serum_samples_od)}")
print(f"Reactive samples (OD > 1.0): {reactive_count}")


# Biological Application 2: Extracting hydrophobic amino acids from a protein fragment
def filter_hydrophobic_residues(sequence):
    """Extracts hydrophobic amino acid single-letter codes (A, V, I, L, M, F, W, P) from a peptide."""
    hydrophobic_set = ['A', 'V', 'I', 'L', 'M', 'F', 'W', 'P']
    matched_residues = []
    
    for aa in sequence.upper():
        if aa in hydrophobic_set:
            matched_residues.append(aa)
            
    return matched_residues

peptide_seq = "ACDEFGHIKLMNPQRSTVWY"
hydrophobic_found = filter_hydrophobic_residues(peptide_seq)

print(f"\nOriginal Peptide: {peptide_seq}")
print(f"Extracted Hydrophobic Residues: {hydrophobic_found}")
print(f"Hydrophobic Fraction: {len(hydrophobic_found) / len(peptide_seq):.2%}")
````

## 6. Functions with Dictionaries

Functions are frequently used to query, transform, or evaluate structured records stored as **dictionaries**. By passing a dictionary into a function, you create a modular parser that can process thousands of records structured with identical key schemes.

### Data Pipeline Architecture

Passing structured metadata dictionaries through custom functions establishes a clean data-processing pipeline:

$$\begin{matrix} \text{Structured Record} \\ \text{(Dictionary)} \end{matrix} \longrightarrow \boxed{\begin{matrix} \text{Query / Filter} \\ \text{Function} \end{matrix}} \longrightarrow \begin{matrix} \text{Extracted Value /} \\ \text{Boolean Flag} \end{matrix} \longrightarrow \begin{matrix} \text{Downstream Analysis} \\ \text{(Conditionals / Plots)} \end{matrix}$$

````python
# Defining a single sample record
sample = {
    "id": "AAT001",
    "species": "T. vivax",
    "host": "cattle",
    "result": "positive"
}

# Function 1: Extracting a specific metadata field
def get_species(sample_record):
    """Extracts and returns the parasite species from a sample dictionary."""
    return sample_record["species"]

# Function 2: Evaluating a condition to return a Boolean flag
def is_positive(sample_record):
    """Returns True if the diagnostic result is positive, otherwise False."""
    return sample_record["result"] == "positive"

# Executing functions on the dictionary record
species = get_species(sample)
status = is_positive(sample)

print(f"Sample ID: {sample['id']}")
print(f"Extracted Species: {species}")
print(f"Is Positive? {status}")
````

### Using Boolean Functions in Downstream Control Flow

Functions that return `True` or `False` (often called **predicate functions**) integrate naturally into conditional logic, allowing you to write highly readable, declarative code.

````python
# Pipelining boolean returns directly into an if statement
if is_positive(sample):
    print(f"ALERT: Positive {get_species(sample)} infection detected in {sample['host']} ({sample['id']}).")
else:
    print(f"Sample {sample['id']} is negative.")
````

### Processing Batches of Dictionaries

Combining **Lists of Dictionaries** with custom functions allows us to filter or summarize large biological metadata sets efficiently.

````python
# List of metadata records representing various disease surveillance samples
sample_batch = [
    {"id": "AAT001", "species": "T. vivax", "host": "cattle", "result": "positive"},
    {"id": "AAT002", "species": "T. congolense", "host": "goat", "result": "negative"},
    {"id": "AAT003", "species": "T. vivax", "host": "cattle", "result": "positive"},
    {"id": "AAT004", "species": "T. brucei", "host": "donkey", "result": "negative"},
    {"id": "AAT005", "species": "T. congolense", "host": "cattle", "result": "positive"}
]

def filter_positive_samples(records):
    """Iterates through a list of sample dictionaries and returns positive records."""
    positive_list = []
    for record in records:
        if is_positive(record):  # Reusing our boolean function inside a loop!
            positive_list.append(record["id"])
    return positive_list

# Extracting positive sample IDs across the batch
positives = filter_positive_samples(sample_batch)
print(f"Positive Sample IDs in Batch: {positives}")
print(f"Infection Rate: {len(positives) / len(sample_batch):.1%}")
````

## 7. Positional and Keyword Arguments

When invoking a function, Python allows us to pass arguments in two distinct ways: by **position** (order matters) or by **keyword** (explicit parameter assignment).

### Key Distinctions

* **Positional Arguments:** Matched strictly by left-to-right order. Fast to write, but can be ambiguous if a function accepts many arguments of the same type.
* **Keyword Arguments (`key=value`):** Matched directly by parameter name. Order becomes irrelevant, making calls self-documenting and significantly easier to read.

````python
# Function definition with 3 parameters
def describe_sample(sample_id, species, host):
    """Formats a human-readable metadata string for a biological sample."""
    return f"{sample_id}: {species} from {host}"

# 1. POSITIONAL CALL (Order-dependent)
# Argument 1 -> sample_id, Argument 2 -> species, Argument 3 -> host
print("Positional Call:")
print(describe_sample("AAT001", "T. vivax", "cattle"))
````

### Using Keyword Arguments

By specifying `parameter_name=value`, we remove ambiguity and break the strict positional constraint.

````python
# 2. KEYWORD CALL (Matching original order)
print("Keyword Call (Standard Order):")
print(
    describe_sample(
        sample_id="AAT001",
        species="T. vivax",
        host="cattle"
    )
)

# 3. KEYWORD CALL (Reordered parameters)
print("\nKeyword Call (Reordered):")
print(
    describe_sample(
        host="cattle",
        species="T. vivax",
        sample_id="AAT001"
    )
)
````

### Investigation: Architectural Readability

Consider calling a scientific function with multiple numeric arguments:

```python
# Hard to interpret without checking the function definition
run_assay(0.05, 37.0, 15, True, False)

# Self-documenting and resilient to parameter order shifts
run_assay(
    absorbance_cutoff=0.05,
    incubation_temp_c=37.0,
    time_minutes=15,
    normalize=True,
    verbose=False
)

* **Critical Rule:** When mixing positional and keyword arguments, positional arguments must ALWAYS come first. Placing a positional argument after a keyword argument causes a `SyntaxError`.

````python
# MIXING POSITIONAL AND KEYWORD ARGUMENTS
# Positional first ('AAT002'), followed by keywords
print(describe_sample("AAT002", host="goat", species="T. congolense"))
````

````python
# Executing the line below causes a SyntaxError:
print(describe_sample(sample_id="AAT002", "T. congolense", host="goat"))
````

### Diagnostic Breakdown: Why Python Enforces This Rule
1. **Positional Ambiguity:** Python evaluates positional arguments left-to-right by filling open parameter slots in the function signature from left to right.
2. **Double Binding Prevention:** If you assign `sample_id="AAT002"` via keyword first, the first slot is filled. If Python then encounters a positional argument `"T. congolense"`, it tries to place it in the first open slot (which was `sample_id`), causing a conflict where a parameter would receive multiple values.
3. **Parsing Efficiency:** Enforcing all positional arguments before any key=value pairs keeps parameter resolution $\mathcal{O}(N)$ deterministic and unambiguous at parse time.

## 8. Default Parameters

Sometimes I want a function to use a sensible value when I do not provide one.

I can assign a default value to a parameter:

```python
def classify_concentration(value, threshold=2.0):
```

Here, `threshold` uses `2.0` unless I provide a different value.

```text
classify_concentration(2.5)
        ↓
threshold = 2.0

classify_concentration(1.8, threshold=1.5)
        ↓
threshold = 1.5
```

A parameter with a default value must come after parameters without defaults in the function definition.

````python
def classify_concentration(value, threshold=2.0):
    if value >= threshold:
        return "high"
    return "normal"


print(classify_concentration(2.5))
print(classify_concentration(1.4))
print(classify_concentration(1.8, threshold=1.5))
````

## 9. Local and Global Scope

Here, `result` is local to `analyze_sample()`.

A variable defined outside a function belongs to the **global scope** and can be accessed from the surrounding code.

## 10. Type Hints

I can add type hints to function parameters and return values to make the expected types clearer.

```python id="7s4k1m"
def calculate_mean(values: list[float]) -> float:
```

Here, `values: list[float]` indicates that I expect a list of numbers, and `-> float` indicates that I expect the function to return a float.

Type hints mainly communicate how I intend the function to be used. They do not automatically convert or validate values at runtime.


````python
def calculate_mean(values: list[float]) -> float:
    total = 0.0

    for value in values:
        total += value

    return total / len(values)


def classify_result(result: str) -> str:
    if result == "positive":
        return "detected"

    return "not detected"


print(calculate_mean([2.0, 4.0, 6.0]))
print(classify_result("positive"))
````

## 11. Functions Working Together

So far, I have used functions to perform individual tasks. I can also combine several functions so that the result from one function becomes input to another.

This allows me to break a larger analysis into smaller steps instead of putting everything into one function.

```text
samples
   │
   ├──→ count_samples() ──────→ total
   │
   └──→ count_positive() ─────→ positive
                                  │
                                  ▼
                         calculate_positive_rate()
                                  │
                                  ▼
                           positive / total

````python
# Exercise 1 Implementations

def calculate_total(data: list[float]) -> float:
    """Return the arithmetic sum of a list of floats."""
    total = 0.0
    for value in data:
        total += value
    return total

def calculate_mean(data: list[float]) -> float:
    """Return the arithmetic mean of a list of floats."""
    if not data:
        return 0.0
    return calculate_total(data) / len(data)

def find_maximum(data: list[float]) -> float:
    """Return the highest numerical value in a list of floats."""
    if not data:
        return 0.0
    
    max_val = data[0]
    for value in data[1:]:
        if value > max_val:
            max_val = value
    return max_val

# --- Validation and Testing ---
concentrations_a = [2.1, 3.4, 1.8, 4.2, 5.1]
concentrations_b = [0.15, 0.82, 0.04, 0.91, 0.33, 0.55]

print("=== Dataset A Results ===")
print("Total:", calculate_total(concentrations_a))
print("Mean: ", calculate_mean(concentrations_a))
print("Max:  ", find_maximum(concentrations_a))

print("\n=== Dataset B Results ===")
print("Total:", round(calculate_total(concentrations_b), 2))
print("Mean: ", round(calculate_mean(concentrations_b), 4))
print("Max:  ", find_maximum(concentrations_b))
````

Use a list of sample dictionaries and build a small analysis from several functions.

Think about:

Sketch the flow of your solution before writing the functions.

````python
# EXERCISE 3 — FUNCTION COMPOSITION

def count_total_samples(samples: list[dict]) -> int:
    """Return the total number of sample records."""
    return len(samples)


def count_positive_samples(samples: list[dict]) -> int:
    """Return the number of samples with a positive result."""
    positive_count = 0

    for sample in samples:
        result = sample.get("result", "").lower()

        if result == "positive":
            positive_count += 1

    return positive_count


def calculate_positive_rate(samples: list[dict]) -> float:
    """Calculate the percentage of samples with a positive result."""
    total = count_total_samples(samples)

    if total == 0:
        return 0.0

    positive = count_positive_samples(samples)

    return (positive / total) * 100


def count_species_distribution(samples: list[dict]) -> dict[str, int]:
    """Return the number of samples recorded for each species."""
    species_counts = {}

    for sample in samples:
        species = sample.get("species", "Unknown Species")

        if species not in species_counts:
            species_counts[species] = 0

        species_counts[species] += 1

    return species_counts


def generate_analysis_summary(samples: list[dict]) -> dict:
    """Combine the results of several analysis functions into one summary."""
    total = count_total_samples(samples)
    positive = count_positive_samples(samples)
    positive_rate = calculate_positive_rate(samples)
    species_distribution = count_species_distribution(samples)

    return {
        "total_samples": total,
        "positive_samples": positive,
        "positivity_rate_pct": round(positive_rate, 2),
        "species_breakdown": species_distribution
    }


# Sample dataset
sample_dataset = [
    {
        "id": "AAT001",
        "species": "T. vivax",
        "result": "positive",
        "antibody_level": 3.7
    },
    {
        "id": "AAT002",
        "species": "T. congolense",
        "result": "negative",
        "antibody_level": 1.2
    },
    {
        "id": "AAT003",
        "species": "T. vivax",
        "result": "positive",
        "antibody_level": 4.1
    },
    {
        "id": "AAT004",
        "species": "T. brucei",
        "result": "negative",
        "antibody_level": 0.8
    },
    {
        "id": "AAT005",
        "species": "T. congolense",
        "result": "positive",
        "antibody_level": 2.9
    }
]


# Generate the analysis
summary = generate_analysis_summary(sample_dataset)


# Display the results
print("=== ANALYSIS SUMMARY REPORT ===")
print(f"Total Samples     : {summary['total_samples']}")
print(f"Positive Samples  : {summary['positive_samples']}")
print(f"Positivity Rate   : {summary['positivity_rate_pct']}%")

print("\nSpecies Frequency Breakdown:")

for species, count in summary["species_breakdown"].items():
    print(f"  {species}: {count} sample(s)")
````

## 14. Integrated Challenge — Research Sample Analyzer

In `05_dictionaries.ipynb`, I built a Research Sample Analyzer using nested dictionaries, loops, conditionals, and manual counting.

I now want to revisit that analyzer and refactor the working logic into functions.

The goal is not to rebuild the analyzer from scratch. I want to identify independent pieces of logic, turn them into reusable functions, and then combine those functions to perform the same analysis.

### Starting Data

```python

### What I Want the Analyzer to Do

The refactored version should allow me to:

1. parse the raw text into a nested dictionary
2. count the total samples
3. count positive and negative results
4. calculate the positivity rate
5. determine the species distribution
6. identify the most frequent species
7. look up an individual sample
8. add a new sample
9. update an existing sample
10. display a summary of the analysis

The functions should receive the data they need through parameters and return results when those results need to be used elsewhere.

### Before Coding

I will first look back at the procedural version from `05_dictionaries.ipynb`.

I will identify:

* Which pieces could be reused?
* Which operations should return values rather than print them?

* Which operations modify the sample dictionary?

````python
# ==============================================================================
# RESEARCH SAMPLE ANALYZER — FUNCTION REFACTORING
# ==============================================================================


def parse_raw_samples(data_str: str) -> dict[str, dict]:
    """Parse raw sample data into a nested dictionary."""
    samples = {}

    rows = data_str.strip().split("\n")

    for row in rows:
        if not row.strip():
            continue

        sample_id, species, host, result = row.strip().split(",")

        samples[sample_id.strip().upper()] = {
            "species": species.strip(),
            "host": host.strip(),
            "result": result.strip().lower()
        }

    return samples


def count_results(
    samples: dict[str, dict]
) -> tuple[int, int, int, float]:
    """Calculate total samples, positive and negative counts, and positivity rate."""
    total = len(samples)
    positives = 0
    negatives = 0

    for record in samples.values():
        result = record["result"]

        if result == "positive":
            positives += 1
        elif result == "negative":
            negatives += 1

    if total > 0:
        positivity_rate = positives / total * 100
    else:
        positivity_rate = 0.0

    return total, positives, negatives, positivity_rate


def get_species_distribution(
    samples: dict[str, dict]
) -> dict[str, int]:
    """Count how many samples belong to each species."""
    counts = {}

    for record in samples.values():
        species = record["species"]
        counts[species] = counts.get(species, 0) + 1

    return counts


def find_most_frequent_species(
    samples: dict[str, dict]
) -> tuple[str | None, int]:
    """Find the species with the highest number of samples."""
    distribution = get_species_distribution(samples)

    top_species = None
    highest_count = 0

    for species, count in distribution.items():
        if count > highest_count:
            highest_count = count
            top_species = species

    return top_species, highest_count


def lookup_sample(
    samples: dict[str, dict],
    sample_id: str
) -> dict | None:
    """Return a sample record by ID, or None if it is not found."""
    clean_id = sample_id.strip().upper()

    return samples.get(clean_id)


def add_sample(
    samples: dict[str, dict],
    sample_id: str,
    species: str,
    host: str,
    result: str
) -> bool:
    """Add a new sample if the sample ID does not already exist."""
    clean_id = sample_id.strip().upper()

    if clean_id in samples:
        return False

    samples[clean_id] = {
        "species": species.strip(),
        "host": host.strip(),
        "result": result.strip().lower()
    }

    return True


def update_sample(
    samples: dict[str, dict],
    sample_id: str,
    species: str,
    host: str,
    result: str
) -> bool:
    """Update an existing sample record."""
    clean_id = sample_id.strip().upper()

    if clean_id not in samples:
        return False

    samples[clean_id] = {
        "species": species.strip(),
        "host": host.strip(),
        "result": result.strip().lower()
    }

    return True


def display_analysis_summary(
    samples: dict[str, dict]
) -> None:
    """Display the main results of the sample analysis."""
    total, positive, negative, positivity_rate = count_results(samples)
    species_distribution = get_species_distribution(samples)
    top_species, top_count = find_most_frequent_species(samples)

    print("=" * 60)
    print(f"{'RESEARCH SAMPLE DATABASE SUMMARY':^60}")
    print("=" * 60)
    print(f"Total Samples Processed : {total:>15d}")
    print(f"Positive Assay Results  : {positive:>15d}")
    print(f"Negative Assay Results  : {negative:>15d}")
    print(f"Positivity Rate         : {positivity_rate:>14.1f}%")
    print("-" * 60)

    print("Species Distribution:")

    for species, count in species_distribution.items():
        print(f"  {species:<22}: {count:>5d}")

    print("-" * 60)
    print(
        f"Most Frequent Species   : "
        f"{top_species} ({top_count} occurrences)"
    )
    print("=" * 60)


# ==============================================================================
# TESTING THE REFACTORED ANALYZER
# ==============================================================================

raw_samples = """AAT001,T. vivax,cattle,positive
AAT002,T. congolense,cattle,negative
AAT003,T. vivax,goat,positive
AAT004,T. vivax,cattle,positive
AAT005,T. congolense,goat,negative"""


# Parse the raw data
db = parse_raw_samples(raw_samples)


# Analyze the original dataset
display_analysis_summary(db)


# Test sample lookup
query_id = "AAT003"
record = lookup_sample(db, query_id)

print(f"\nLookup: {query_id}")
print(record)


# Test adding a new sample
added = add_sample(
    db,
    "AAT006",
    "T. brucei",
    "human",
    "positive"
)

print(f"\nAAT006 added: {added}")


# Test updating an existing sample
updated = update_sample(
    db,
    "AAT002",
    "T. congolense",
    "cattle",
    "positive"
)

print(f"AAT002 updated: {updated}")


# Analyze the modified dataset
print("\nAfter adding AAT006 and updating AAT002:\n")
display_analysis_summary(db)
````

Python maps positional arguments from left to right:

```text id="qj2m8w"
sample_id ← "T. vivax"
species   ← "AAT001"
```

The function therefore produces an output that is syntactically valid but represents the data incorrectly.

````python
# Corect code
result = describe_sample("AAT001", "T. vivax")

print(result)
````

````python
# Or just make the argument names explicit
result = describe_sample(
    sample_id="AAT001",
    species="T. vivax"
)

print(result)
````

### What I Observed

This problem did not produce an error message. The code ran successfully, but the result was wrong.

This is an important debugging distinction: **code can execute without an exception and still contain a logic error.**

When a function produces unexpected output, I should not only look for error messages. I also need to check whether the values were passed to the correct parameters.

### Debugging Exercise 4 — Local Variable and Scope

#### Broken Code

```python id="z8q4wm"
def calculate_total(values):
    total = sum(values)
    return total

calculate_total([1, 2, 3])

print(total)
```

### Before Running

I want to trace where `total` is created and where I am trying to use it.

* Where is `total` defined?
* Where can the function access `total`?
* Is `total` available outside the function?
* What value does the function return?
* What happens if I call the function without storing its returned value?
* What type of error do I expect?

### My Prediction

`total` is created inside `calculate_total()`, so it belongs to the function's local scope.

The function returns the value:

```text id="q8v4rm"
6
```

However, I am not storing that returned value anywhere:

```python id="m7w2ks"
calculate_total([1, 2, 3])
```

The variable `total` itself is not created in the surrounding code.

I therefore expect `print(total)` to raise a `NameError`.

````python
# Run the code
def calculate_total(values):
    total = sum(values)
    return total

calculate_total([1, 2, 3])

print(total)
````

### Diagnose the Problem

`total` is a **local variable** because it is created inside the function.

It can be used here:

```python id="v3j9nd"
def calculate_total(values):
    total = sum(values)
    return total
```

But it cannot be accessed directly from outside the function:

```python id="a2k6wp"
print(total)
```

The function already provides a way to make the calculated value available outside its local scope: **`return`**.

The problem is that I called the function but ignored the returned value.

````python
# Correct Code

# Capture the returned value in a variable outside the function:
def calculate_total(values):
    total = sum(values)
    return total

grand_total = calculate_total([1, 2, 3])

print(grand_total)
````

### What I Observed

The variable `total` belongs to the function's local scope. I cannot use that variable directly outside the function.

What I *can* use outside the function is the value returned by the function:

```text id="h4r7vz"
local variable
     ↓
   return
     ↓
calling code
     ↓
grand_total
```

This connects directly to the previous debugging exercise. A function can calculate a value correctly, but I need to **return it and capture that returned value** if I want to use it elsewhere.

When I see a `NameError` involving a variable created inside a function, I should check whether I am trying to access a local variable directly instead of using the function's returned value.
